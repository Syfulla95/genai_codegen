```python
# ======================================================================
# ORIGINAL PYSPARK CODE
# ======================================================================


# ============================================================================
# AI GENERATED PYSPARK CODE - ISSUE GA-23942
# Processing Mode: Individual Node Processing
# ============================================================================

# Import required libraries
from pyspark.sql import functions as F
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

def log_df_info(df, step_name):
    """Log DataFrame information for debugging"""
    logger.info(f"{step_name}: {df.count()} rows, {len(df.columns)} columns")
    logger.info(f"Schema: {[f.name + ':' + str(f.dataType) for f in df.schema.fields]}")


# ============================================================================
# app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# ============================================================================
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_0: $NDW_XT.BIL_AccountStatusType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table containing account status type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_0: $NDW_XT.BIL_AccountStatusType_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_accountstatustype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_AccountStatusType_Id_Xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_0 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_0, "Source Node node_0: $NDW_XT.BIL_AccountStatusType_Id_Xt")
    
    logger.info("✓ Node node_0 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_0 ($NDW_XT.BIL_AccountStatusType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_0: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_1: $ndw_type.AccountStatusType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for checking existing AccountStatusType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_1: $ndw_type.AccountStatusType")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_type.AccountStatusType"
    local_path = "/mnt/data/ndw_type/AccountStatusType.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_1 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_1, "Source Node node_1: $ndw_type.AccountStatusType")
    
    logger.info("✓ Node node_1 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_1 ($ndw_type.AccountStatusType): {str(e)}")
    raise Exception(f"Critical error in Node node_1: {str(e)}")
# ============================================================================
# NODE node_2: $WORKDB_BILLING.TW6C0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_2: $WORKDB_BILLING.TW6C0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw6c0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw6c0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_2 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6C0_INCR - Node node_2 ===")
    log_df_info(final_df_node_2, "Source DataFrame Information")
    
    logger.info("✓ Node node_2 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_2 ($WORKDB_BILLING.TW6C0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_2: {str(e)}")
# ============================================================================
# NODE node_3: $NDW_XT.BIL_CommentType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to find the maximum CommentType_Id and filter existing Comments_Code.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_3: $NDW_XT.BIL_CommentType_Id_XT (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_CommentType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_CommentType_Id_XT.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_3 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CommentType_Id_XT - Node node_3 ===")
    log_df_info(final_df_node_3, "Source Node node_3")
    
    logger.info("✓ Node node_3 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_3 ($NDW_XT.BIL_CommentType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_3: {str(e)}")
# ============================================================================
# NODE node_4: $NDW_XT.BIL_CommentType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing comment type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_4: $NDW_XT.BIL_CommentType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_commenttype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_CommentType_Id_XT.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_4 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CommentType_Id_XT - Node node_4 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_4.columns)}")
    final_df_node_4.printSchema()
    final_df_node_4.show(3, truncate=False)
    
    logger.info("✓ Node node_4 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_4 ($NDW_XT.BIL_CommentType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_4: {str(e)}")
# ============================================================================
# NODE node_5: $ndw_type.CommentType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing CommentType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_5: $ndw_type.CommentType (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "CommentType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/ndw_type/CommentType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_5 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.CommentType - Node node_5 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_5.columns)}")
    final_df_node_5.printSchema()
    final_df_node_5.show(3, truncate=False)
    
    logger.info("✓ Node node_5 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_5 ($ndw_type.CommentType): {str(e)}")
    raise Exception(f"Critical error in Node node_5: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_6: $WORKDB_BILLING.TA205_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental transaction codes to be processed.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_6: $WORKDB_BILLING.TA205_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_ta205_incr"
    local_path = "/mnt/data/workdb_billing_ta205_incr.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_6 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_6, "Source Node node_6: $WORKDB_BILLING.TA205_INCR")
    
    logger.info("✓ Node node_6 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_6 ($WORKDB_BILLING.TA205_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_6: {str(e)}")
# ============================================================================
# NODE node_7: $NDW_XT.BIL_DiaryTransactionType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum DiaryTransactionType_Id and check for existing transaction codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_7: $NDW_XT.BIL_DiaryTransactionType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_DiaryTransactionType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_DiaryTransactionType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_7 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DiaryTransactionType_Id_XT - Node node_7 ===")
    log_df_info(final_df_node_7, "Source DataFrame")
    
    logger.info("✓ Node node_7 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_7 ($NDW_XT.BIL_DiaryTransactionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_7: {str(e)}")
# ============================================================================
# NODE node_8: $NDW_XT.BIL_DiaryTransactionType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing DiaryTransactionType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_8: $NDW_XT.BIL_DiaryTransactionType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_diarytransactiontype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_xt_bil_diarytransactiontype_id_xt.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_8 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DiaryTransactionType_Id_XT - Node node_8 ===")
    log_df_info(final_df_node_8, "Source Data Loaded")
    
    logger.info("✓ Node node_8 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_8 ($NDW_XT.BIL_DiaryTransactionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_8: {str(e)}")
# ============================================================================
# NODE node_9: $ndw_type.DiaryTransactionType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing DiaryTransactionType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_9: $ndw_type.DiaryTransactionType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "DiaryTransactionType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_9 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.DiaryTransactionType - Node node_9 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_9.columns)}")
    final_df_node_9.printSchema()
    final_df_node_9.show(3, truncate=False)
    
    logger.info("✓ Node node_9 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_9 ($ndw_type.DiaryTransactionType): {str(e)}")
    raise Exception(f"Critical error in Node node_9: {str(e)}")
# ============================================================================
# NODE node_10: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_10: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper
    final_df_node_10 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_10 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_10.columns)}")
    final_df_node_10.printSchema()
    final_df_node_10.show(3, truncate=False)
    
    logger.info("✓ Node node_10 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_10 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_10: {str(e)}")
# ============================================================================
# NODE node_11: $NDW_XT.BIL_FinanTransStatusType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to check existing Status_Code values and calculate max FinancialTransStatusType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_11: $NDW_XT.BIL_FinanTransStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_finantransstatustype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_11 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_FinanTransStatusType_Id_XT - Node node_11 ===")
    log_df_info(final_df_node_11, "Source DataFrame Information")
    
    logger.info("✓ Node node_11 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_11 ($NDW_XT.BIL_FinanTransStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_11: {str(e)}")
# ============================================================================
# NODE node_12: $NDW_XT.BIL_FinanTransStatusType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing financial transaction status type data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_12: $NDW_XT.BIL_FinanTransStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_FinanTransStatusType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper
    final_df_node_12 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_FinanTransStatusType_Id_XT - Node node_12 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_12.columns)}")
    final_df_node_12.printSchema()
    final_df_node_12.show(3, truncate=False)
    
    logger.info("✓ Node node_12 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_12 ($NDW_XT.BIL_FinanTransStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_12: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_13: $ndw_type.FinancialTransStatusType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for comparison to filter out existing records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_13: $ndw_type.FinancialTransStatusType")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_type.FinancialTransStatusType"
    local_path = "/mnt/data/ndw_type_FinancialTransStatusType.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_13 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_13, "Source Node node_13: $ndw_type.FinancialTransStatusType")
    
    logger.info("✓ Node node_13 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_13 ($ndw_type.FinancialTransStatusType): {str(e)}")
    raise Exception(f"Critical error in Node node_13: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_14: $WORKDB_BILLING.TZ5Z2_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental Charge_Type data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_14: $WORKDB_BILLING.TZ5Z2_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz5z2_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5Z2_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_14 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_14, "Source Node node_14: $WORKDB_BILLING.TZ5Z2_INCR")
    
    logger.info("✓ Node node_14 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_14 ($WORKDB_BILLING.TZ5Z2_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_14: {str(e)}")
# ============================================================================
# NODE node_15: $NDW_XT.BIL_InstallmentFeeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum InstallmentFeeType_Id and filter existing Charge_Type values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_15: $NDW_XT.BIL_InstallmentFeeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai_powerbi_catalog.ndw_xt.bil_installmentfeetype_id_xt"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/ndw_xt_bil_installmentfeetype_id_xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_15 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_InstallmentFeeType_Id_XT - Node node_15 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_15.columns)}")
    final_df_node_15.printSchema()
    final_df_node_15.show(3, truncate=False)
    
    logger.info("✓ Node node_15 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_15 ($NDW_XT.BIL_InstallmentFeeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_15: {str(e)}")
# ============================================================================
# NODE node_16: $NDW_XT.BIL_InstallmentFeeType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing installment fee type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_16: $NDW_XT.BIL_InstallmentFeeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_installmentfeetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_InstallmentFeeType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_16 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_InstallmentFeeType_Id_XT - Node node_16 ===")
    log_df_info(final_df_node_16, "Source DataFrame Information")
    
    logger.info("✓ Node node_16 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_16 ($NDW_XT.BIL_InstallmentFeeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_16: {str(e)}")
# ============================================================================
# NODE node_17: $ndw_type.InstallmentFeeType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for checking existing InstallmentFeeType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_17: $ndw_type.InstallmentFeeType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "InstallmentFeeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/InstallmentFeeType.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_17 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.InstallmentFeeType - Node node_17 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_17.columns)}")
    final_df_node_17.printSchema()
    final_df_node_17.show(3, truncate=False)
    
    logger.info("✓ Node node_17 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_17 ($ndw_type.InstallmentFeeType): {str(e)}")
    raise Exception(f"Critical error in Node node_17: {str(e)}")
# ============================================================================
# NODE node_18: $WORKDB_BILLING.TW5T2_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_18: $WORKDB_BILLING.TW5T2_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "TW5T2_INCR"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_18 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T2_INCR - Node node_18 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_18.columns)}")
    final_df_node_18.printSchema()
    final_df_node_18.show(3, truncate=False)
    
    logger.info("✓ Node node_18 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_18 ($WORKDB_BILLING.TW5T2_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_18: {str(e)}")
# ============================================================================
# NODE node_19: $NDW_XT.BIL_ProcessingSourceType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum ProcessingSourceType_Id and check for existing Bank_Account_Code.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_19: $NDW_XT.BIL_ProcessingSourceType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProcessingSourceType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProcessingSourceType_Id_Xt.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_19 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ProcessingSourceType_Id_Xt - Node node_19 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_19.columns)}")
    final_df_node_19.printSchema()
    final_df_node_19.show(3, truncate=False)
    
    logger.info("✓ Node node_19 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_19 ($NDW_XT.BIL_ProcessingSourceType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_19: {str(e)}")
# ============================================================================
# NODE node_20: $WORKDB_BILLING.TW5T9_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_20: $WORKDB_BILLING.TW5T9_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5t9_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5t9_incr.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_20 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T9_INCR - Node node_20 ===")
    log_df_info(final_df_node_20, "Source DataFrame Information")
    
    logger.info("✓ Node node_20 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_20 ($WORKDB_BILLING.TW5T9_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_20: {str(e)}")
# ============================================================================
# NODE node_21: $NDW_XT.BIL_ProcessingSourceType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum ProcessingSourceType_Id and filter existing Bank_Account_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_21: $NDW_XT.BIL_ProcessingSourceType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProcessingSourceType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProcessingSourceType_Id_Xt.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_21 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ProcessingSourceType_Id_Xt - Node node_21 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_21.columns)}")
    final_df_node_21.printSchema()
    final_df_node_21.show(3, truncate=False)
    
    logger.info("✓ Node node_21 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_21 ($NDW_XT.BIL_ProcessingSourceType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_21: {str(e)}")
# ============================================================================
# NODE node_22: $NDW_XT.BIL_ProcessingSourceType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing processing source type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_22: $NDW_XT.BIL_ProcessingSourceType_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProcessingSourceType_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_ProcessingSourceType_Id_Xt.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_22 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ProcessingSourceType_Id_Xt - Node node_22 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_22.columns)}")
    final_df_node_22.printSchema()
    final_df_node_22.show(3, truncate=False)
    
    logger.info("✓ Node node_22 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_22 ($NDW_XT.BIL_ProcessingSourceType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_22: {str(e)}")
# ============================================================================
# NODE node_23: $ndw_type.ProcessingSourceType (Source)
# Container: Batch 1 (Active)
# Description: Target table used to check for existing ProcessingSourceType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_23: $ndw_type.ProcessingSourceType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "app_BLBIS_BILL3"
    source_table = "BLBIS_BIL_Id_XT1_Analysis"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/app_BLBIS_BILL3/BLBIS_BIL_Id_XT1_Analysis.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_23 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ProcessingSourceType - Node node_23 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_23.columns)}")
    final_df_node_23.printSchema()
    final_df_node_23.show(3, truncate=False)
    
    logger.info("✓ Node node_23 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_23 ($ndw_type.ProcessingSourceType): {str(e)}")
    raise Exception(f"Critical error in Node node_23: {str(e)}")
# ============================================================================
# NODE node_24: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_24: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tz4a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_24 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_24 ===")
    log_df_info(final_df_node_24, "Source DataFrame Information")
    
    logger.info("✓ Node node_24 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_24 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_24: {str(e)}")
# ============================================================================
# NODE node_25: $NDW_XT.BIL_RegionType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum RegionType_Id and filter existing Tokenized_RO values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_25: $NDW_XT.BIL_RegionType_Id_XT (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_RegionType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_RegionType_Id_XT.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_25 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_RegionType_Id_XT - Node node_25 ===")
    log_df_info(final_df_node_25, "Source DataFrame Information")
    
    logger.info("✓ Node node_25 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_25 ($NDW_XT.BIL_RegionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_25: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_26: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_26: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_26 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_26, "Source Node node_26: $WORKDB_BILLING.TZ4A0_INCR")
    
    logger.info("✓ Node node_26 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_26 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_26: {str(e)}")
# ============================================================================
# NODE node_27: $NDW_XT.BIL_RegionType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve existing RegionType_Id and Tokenized_RO values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_27: $NDW_XT.BIL_RegionType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_RegionType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_RegionType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_27 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_RegionType_Id_XT - Node node_27 ===")
    log_df_info(final_df_node_27, "Source DataFrame Information")
    
    logger.info("✓ Node node_27 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_27 ($NDW_XT.BIL_RegionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_27: {str(e)}")
# ============================================================================
# NODE node_28: $WORKDB_BILLING.TW5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_28: $WORKDB_BILLING.TW5A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5A0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5A0_INCR.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper
    final_df_node_28 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5A0_INCR - Node node_28 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_28.columns)}")
    final_df_node_28.printSchema()
    final_df_node_28.show(3, truncate=False)
    
    logger.info("✓ Node node_28 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_28 ($WORKDB_BILLING.TW5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_28: {str(e)}")
# ============================================================================
# NODE node_29: $NDW_XT.BIL_RegionType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to check existing regions and calculate max RegionType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_29: $NDW_XT.BIL_RegionType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_RegionType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_RegionType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_29 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_RegionType_Id_XT - Node node_29 ===")
    log_df_info(final_df_node_29, "Source DataFrame Information")
    
    logger.info("✓ Node node_29 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_29 ($NDW_XT.BIL_RegionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_29: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_30: $NDW_XT.BIL_RegionType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing RegionType_Id and Region data for insertion.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_30: $NDW_XT.BIL_RegionType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_regiontype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_RegionType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_30 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_30, "Source Node node_30: $NDW_XT.BIL_RegionType_Id_XT")
    
    logger.info("✓ Node node_30 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_30 ($NDW_XT.BIL_RegionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_30: {str(e)}")
# ============================================================================
# NODE node_31: $ndw_type.RegionType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing RegionType_Id values and collecting statistics.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_31: $ndw_type.RegionType (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "RegionType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for local testing)
    local_file_path = "/mnt/data/ndw_type_regiontype.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_31 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.RegionType - Node node_31 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_31.columns)}")
    final_df_node_31.printSchema()
    final_df_node_31.show(3, truncate=False)
    
    logger.info("✓ Node node_31 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_31 ($ndw_type.RegionType): {str(e)}")
    raise Exception(f"Critical error in Node node_31: {str(e)}")
# ============================================================================
# NODE node_32: $WORKDB_BILLING.TZ5BO_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_32: $WORKDB_BILLING.TZ5BO_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tz5bo_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tz5bo_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_32 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ5BO_INCR - Node node_32 ===")
    log_df_info(final_df_node_32, "Source DataFrame Information")
    
    logger.info("✓ Node node_32 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_32 ($WORKDB_BILLING.TZ5BO_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_32: {str(e)}")
# ============================================================================
# NODE node_33: $NDW_XT.BIL_RegistrationStatType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum RegistrationStatusType_Id and filter existing Agent_Active_Flag values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_33: $NDW_XT.BIL_RegistrationStatType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_RegistrationStatType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_RegistrationStatType_Id_Xt.csv"
    
    # Read data using the universal read_table helper
    final_df_node_33 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_RegistrationStatType_Id_Xt - Node node_33 ===")
    log_df_info(final_df_node_33, "Source DataFrame Information")
    
    logger.info("✓ Node node_33 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_33 ($NDW_XT.BIL_RegistrationStatType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_33: {str(e)}")
# ============================================================================
# NODE node_34: $NDW_XT.BIL_RegistrationStatType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table containing registration status type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_34: $NDW_XT.BIL_RegistrationStatType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_RegistrationStatType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_RegistrationStatType_Id_Xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_34 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_RegistrationStatType_Id_Xt - Node node_34 ===")
    log_df_info(final_df_node_34, "Source DataFrame Information")
    
    logger.info("✓ Node node_34 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_34 ($NDW_XT.BIL_RegistrationStatType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_34: {str(e)}")
# ============================================================================
# NODE node_35: $ndw_type.RegistrationStatusType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing RegistrationStatusType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_35: $ndw_type.RegistrationStatusType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "RegistrationStatusType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_35 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.RegistrationStatusType - Node node_35 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_35.columns)}")
    final_df_node_35.printSchema()
    final_df_node_35.show(3, truncate=False)
    
    logger.info("✓ Node node_35 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_35 ($ndw_type.RegistrationStatusType): {str(e)}")
    raise Exception(f"Critical error in Node node_35: {str(e)}")
# ============================================================================
# NODE node_36: $WORKDB_BILLING.TW5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Account Codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_36: $WORKDB_BILLING.TW5A0_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_36 = read_table(
        table_name=full_table_name,
        local_file_path=None  # No fallback file path provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5A0_INCR - Node node_36 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_36.columns)}")
    final_df_node_36.printSchema()
    final_df_node_36.show(3, truncate=False)
    
    logger.info("✓ Node node_36 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_36 ($WORKDB_BILLING.TW5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_36: {str(e)}")
# ============================================================================
# NODE node_37: $NDW_XT.BIL_AccountCodeType_Id_Xt (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to determine the maximum AccountCodeType_Id and filter existing Account Codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_37: $NDW_XT.BIL_AccountCodeType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AccountCodeType_Id_Xt"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_AccountCodeType_Id_Xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_37 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_AccountCodeType_Id_Xt - Node node_37 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_37.columns)}")
    final_df_node_37.printSchema()
    final_df_node_37.show(3, truncate=False)
    
    logger.info("✓ Node node_37 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_37 ($NDW_XT.BIL_AccountCodeType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_37: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv") \
                .option("header", "true") \
                .option("inferSchema", "true") \
                .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================
# NODE node_38: $NDW_XT.BIL_AccountCodeType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table containing account code type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_38: $NDW_XT.BIL_AccountCodeType_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_accountcodetype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_AccountCodeType_Id_Xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_38 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_38, "Source Node node_38: $NDW_XT.BIL_AccountCodeType_Id_Xt")
    
    logger.info("✓ Node node_38 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_38 ($NDW_XT.BIL_AccountCodeType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_38: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_39: $ndw_type.AccountCodeType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for checking existing account code type IDs and for inserting new records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_39: $ndw_type.AccountCodeType")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_type.AccountCodeType"
    local_path = "/mnt/data/ndw_type/AccountCodeType.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_39 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_39, "Source Node node_39: $ndw_type.AccountCodeType")
    
    logger.info("✓ Node node_39 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_39 ($ndw_type.AccountCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_39: {str(e)}")
# ============================================================================
# NODE node_40: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Policy Account Status.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_40: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tz4a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_40 = read_table(
        table_name=full_table_name,
        local_file_path=None  # No fallback file path provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_40 ===")
    log_df_info(final_df_node_40, step_name="Source Data Loaded")
    
    logger.info("✓ Node node_40 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_40 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_40: {str(e)}")
# ============================================================================
# NODE node_41: $NDW_XT.BIL_AgreementStatusType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum AgreementStatusType_Id for calculation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_41: $NDW_XT.BIL_AgreementStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementStatusType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_41 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_AgreementStatusType_Id_XT - Node node_41 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_41.columns)}")
    final_df_node_41.printSchema()
    final_df_node_41.show(3, truncate=False)
    
    logger.info("✓ Node node_41 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_41 ($NDW_XT.BIL_AgreementStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_41: {str(e)}")
# ============================================================================
# NODE node_42: $NDW_XT.BIL_AgreementStatusType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing Agreement Status Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_42: $NDW_XT.BIL_AgreementStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "default"
    source_table = "bil_agreementstatustype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"
    
    # Read data using the universal read_table helper
    final_df_node_42 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_AgreementStatusType_Id_XT - Node node_42 ===")
    log_df_info(final_df_node_42, "Source Data Loaded")
    
    logger.info("✓ Node node_42 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_42 ($NDW_XT.BIL_AgreementStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_42: {str(e)}")
# ============================================================================
# NODE node_43: $ndw_type.AgreementStatusType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for checking existing AgreementStatusType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_43: $ndw_type.AgreementStatusType")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai_powerbi_catalog.ndw_type.AgreementStatusType"
    local_path = "/mnt/data/ndw_type/AgreementStatusType.csv"  # Example fallback path
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_43 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.AgreementStatusType - Node node_43 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_43.columns)}")
    final_df_node_43.printSchema()
    final_df_node_43.show(3, truncate=False)
    
    logger.info("✓ Node node_43 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_43 ($ndw_type.AgreementStatusType): {str(e)}")
    raise Exception(f"Critical error in Node node_43: {str(e)}")
# ============================================================================
# NODE node_44: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_44: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"  # Fallback local file path
    
    # Read data using the universal read_table helper function
    final_df_node_44 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_44 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_44.columns)}")
    final_df_node_44.printSchema()
    final_df_node_44.show(3, truncate=False)
    
    logger.info("✓ Node node_44 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_44 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_44: {str(e)}")
# ============================================================================
# NODE node_45: $NDW_XT.BIL_AgreementType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum AgreementType_Id and filter existing External_System values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_45: $NDW_XT.BIL_AgreementType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementType_Id_XT"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_45 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_AgreementType_Id_XT - Node node_45 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_45.columns)}")
    final_df_node_45.printSchema()
    final_df_node_45.show(3, truncate=False)
    
    logger.info("✓ Node node_45 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_45 ($NDW_XT.BIL_AgreementType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_45: {str(e)}")
# ============================================================================
# NODE node_46: $NDW_XT.BIL_AgreementType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing agreement type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_46: $NDW_XT.BIL_AgreementType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_AgreementType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_46 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_AgreementType_Id_XT - Node node_46 ===")
    log_df_info(final_df_node_46, "Source DataFrame Information")
    
    logger.info("✓ Node node_46 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_46 ($NDW_XT.BIL_AgreementType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_46: {str(e)}")
# ============================================================================
# NODE node_47: $ndw_type.AgreementType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing agreement type IDs and for inserting new records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_47: $ndw_type.AgreementType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "AgreementType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type_agreementtype.csv"  # Example path, adjust as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_47 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.AgreementType - Node node_47 ===")
    log_df_info(final_df_node_47, "Source DataFrame Information")
    
    logger.info("✓ Node node_47 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_47 ($ndw_type.AgreementType): {str(e)}")
    raise Exception(f"Critical error in Node node_47: {str(e)}")
# ============================================================================
# NODE node_48: $WORKDB_BILLING.TW5JB_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_48: $WORKDB_BILLING.TW5JB_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5JB_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5JB_INCR.csv"  # Fallback path for local file
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_48 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5JB_INCR - Node node_48 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_48.columns)}")
    final_df_node_48.printSchema()
    final_df_node_48.show(3, truncate=False)
    
    logger.info("✓ Node node_48 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_48 ($WORKDB_BILLING.TW5JB_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_48: {str(e)}")
# ============================================================================
# NODE node_49: $NDW_XT.BIL_BankAccountType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum BankAccountType_Id and filter existing account types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_49: $NDW_XT.BIL_BankAccountType_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BankAccountType_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_BankAccountType_Id_Xt.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper
    final_df_node_49 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BankAccountType_Id_Xt - Node node_49 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_49.columns)}")
    final_df_node_49.printSchema()
    final_df_node_49.show(3, truncate=False)
    
    logger.info("✓ Node node_49 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_49 ($NDW_XT.BIL_BankAccountType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_49: {str(e)}")
# ============================================================================
# NODE node_50: $WORKDB_BILLING.TU5P0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_50: $WORKDB_BILLING.TU5P0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.workdb_billing_tu5p0_incr"
    local_path = "/mnt/data/workdb_billing_tu5p0_incr.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper
    final_df_node_50 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TU5P0_INCR - Node node_50 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_50.columns)}")
    final_df_node_50.printSchema()
    final_df_node_50.show(3, truncate=False)
    
    logger.info("✓ Node node_50 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_50 ($WORKDB_BILLING.TU5P0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_50: {str(e)}")
# ============================================================================
# NODE node_51: $NDW_XT.BIL_BankAccountType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing existing bank account type IDs.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_51: $NDW_XT.BIL_BankAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BankAccountType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_BankAccountType_Id_XT.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_51 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BankAccountType_Id_XT - Node node_51 ===")
    log_df_info(final_df_node_51, "Source DataFrame Information")
    
    logger.info("✓ Node node_51 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_51 ($NDW_XT.BIL_BankAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_51: {str(e)}")
# ============================================================================
# NODE node_52: $NDW_XT.BIL_BankAccountType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing bank account type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_52: $NDW_XT.BIL_BankAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_BankAccountType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_BankAccountType_Id_XT.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_52 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BankAccountType_Id_XT - Node node_52 ===")
    log_df_info(final_df_node_52, "Source Node node_52")
    
    logger.info("✓ Node node_52 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_52 ($NDW_XT.BIL_BankAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_52: {str(e)}")
# ============================================================================
# NODE node_53: $ndw_type.BankAccountType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing BankAccountType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_53: $ndw_type.BankAccountType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "BankAccountType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_53 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.BankAccountType - Node node_53 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_53.columns)}")
    final_df_node_53.printSchema()
    final_df_node_53.show(3, truncate=False)
    
    logger.info("✓ Node node_53 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_53 ($ndw_type.BankAccountType): {str(e)}")
    raise Exception(f"Critical error in Node node_53: {str(e)}")
# ============================================================================
# NODE node_54: $WORKDB_BILLING.TA205_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Diary_Code.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_54: $WORKDB_BILLING.TA205_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "ta205_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/ta205_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_54 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TA205_INCR - Node node_54 ===")
    log_df_info(final_df_node_54, "Source Data Loaded")
    
    logger.info("✓ Node node_54 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_54 ($WORKDB_BILLING.TA205_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_54: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_55: $NDW_XT.BIL_BillingDiaryType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum BillingDiaryType_Id and check existing Diary_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_55: $NDW_XT.BIL_BillingDiaryType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_billingdiarytype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_BillingDiaryType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_55 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_55, "Source Node node_55: $NDW_XT.BIL_BillingDiaryType_Id_XT")
    
    logger.info("✓ Node node_55 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_55 ($NDW_XT.BIL_BillingDiaryType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_55: {str(e)}")
# ============================================================================
# NODE node_56: $NDW_XT.BIL_BillingDiaryType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing BillingDiaryType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_56: $NDW_XT.BIL_BillingDiaryType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_BillingDiaryType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_BillingDiaryType_Id_XT.csv"
    
    # Read data using the universal read_table helper
    final_df_node_56 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BillingDiaryType_Id_XT - Node node_56 ===")
    log_df_info(final_df_node_56, "Source DataFrame Information")
    
    logger.info("✓ Node node_56 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_56 ($NDW_XT.BIL_BillingDiaryType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_56: {str(e)}")
# ============================================================================
# NODE node_57: $ndw_type.BillingDiaryType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for comparison to ensure no duplicate entries are inserted.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_57: $ndw_type.BillingDiaryType (Source)")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.BillingDiaryType"
    local_path = "/mnt/data/BillingDiaryType.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_57 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.BillingDiaryType - Node node_57 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_57.columns)}")
    final_df_node_57.printSchema()
    final_df_node_57.show(3, truncate=False)
    
    logger.info("✓ Node node_57 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_57 ($ndw_type.BillingDiaryType): {str(e)}")
    raise Exception(f"Critical error in Node node_57: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_58: $WORKDB_BILLING.TZ5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for account cancellation reasons.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_58: $WORKDB_BILLING.TZ5A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_58 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_58, "Source Node node_58: $WORKDB_BILLING.TZ5A0_INCR")
    
    logger.info("✓ Node node_58 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_58 ($WORKDB_BILLING.TZ5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_58: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_59: $NDW_XT.BIL_CancelReasonType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum CancelReasonType_Id and filter existing account cancellation reasons.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_59: $NDW_XT.BIL_CancelReasonType_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_cancelreasontype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_CancelReasonType_Id_Xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_59 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_59, "Source Node node_59: $NDW_XT.BIL_CancelReasonType_Id_Xt")
    
    logger.info("✓ Node node_59 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_59 ($NDW_XT.BIL_CancelReasonType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_59: {str(e)}")
# ============================================================================
# NODE node_60: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_60: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_60 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_60 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_60.columns)}")
    final_df_node_60.printSchema()
    final_df_node_60.show(3, truncate=False)
    
    logger.info("✓ Node node_60 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_60 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_60: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_61: $NDW_XT.BIL_CancelReasonType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum CancelReasonType_Id and filter existing Acct_Cancel_Reason values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_61: $NDW_XT.BIL_CancelReasonType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_cancelreasontype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_cancelreasontype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_61 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_61, "Source Node node_61: $NDW_XT.BIL_CancelReasonType_Id_XT")
    
    logger.info("✓ Node node_61 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_61 ($NDW_XT.BIL_CancelReasonType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_61: {str(e)}")
# ============================================================================
# NODE node_62: $WORKDB_BILLING.TZ4WN_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_62: $WORKDB_BILLING.TZ4WN_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "app_BLBIS_BILL3"
    source_table = "BLBIS_BIL_Id_XT1_Analysis"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TZ4WN_INCR.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_62 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4WN_INCR - Node node_62 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_62.columns)}")
    final_df_node_62.printSchema()
    final_df_node_62.show(3, truncate=False)
    
    logger.info("✓ Node node_62 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_62 ($WORKDB_BILLING.TZ4WN_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_62: {str(e)}")
# ============================================================================
# NODE node_63: $NDW_XT.BIL_CancelReasonType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to check existing cancel reasons and retrieve maximum CancelReasonType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_63: $NDW_XT.BIL_CancelReasonType_Id_XT")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_CancelReasonType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_CancelReasonType_Id_XT.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper
    final_df_node_63 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CancelReasonType_Id_XT - Node node_63 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_63.columns)}")
    final_df_node_63.printSchema()
    final_df_node_63.show(3, truncate=False)
    
    logger.info("✓ Node node_63 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_63 ($NDW_XT.BIL_CancelReasonType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_63: {str(e)}")
# ============================================================================
# NODE node_64: $NDW_XT.BIL_CancelReasonType_Id_Xt (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Source table containing cancel reason type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_64: $NDW_XT.BIL_CancelReasonType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_CancelReasonType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_CancelReasonType_Id_Xt.csv"  # Adjust path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_64 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CancelReasonType_Id_Xt - Node node_64 ===")
    log_df_info(final_df_node_64, "Source DataFrame Information")
    
    logger.info("✓ Node node_64 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_64 ($NDW_XT.BIL_CancelReasonType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_64: {str(e)}")
# ============================================================================
# NODE node_65: $ndw_type.CancelReasonType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used to check for existing CancelReasonType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_65: $ndw_type.CancelReasonType")
    
    # Define the Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.CancelReasonType"
    local_path = "/mnt/data/CancelReasonType.csv"  # Example fallback path
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_65 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.CancelReasonType - Node node_65 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_65.columns)}")
    final_df_node_65.printSchema()
    final_df_node_65.show(3, truncate=False)
    
    logger.info("✓ Node node_65 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_65 ($ndw_type.CancelReasonType): {str(e)}")
    raise Exception(f"Critical error in Node node_65: {str(e)}")
# ============================================================================
# NODE node_66: $WORKDB_BILLING.TW5T3_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_66: $WORKDB_BILLING.TW5T3_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "default"
    table_name = "TW5T3_INCR"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5T3_INCR.csv"  # Update path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_66 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T3_INCR - Node node_66 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_66.columns)}")
    final_df_node_66.printSchema()
    final_df_node_66.show(3, truncate=False)
    
    logger.info("✓ Node node_66 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_66 ($WORKDB_BILLING.TW5T3_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_66: {str(e)}")
# ============================================================================
# NODE node_67: $NDW_XT.BIL_ChargeItemType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum ChargeItemType_Id for calculation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_67: $NDW_XT.BIL_ChargeItemType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_chargeitemtype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_xt_bil_chargeitemtype_id_xt.csv"  # Example path
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_67 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ChargeItemType_Id_XT - Node node_67 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_67.columns)}")
    final_df_node_67.printSchema()
    final_df_node_67.show(3, truncate=False)
    
    logger.info("✓ Node node_67 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_67 ($NDW_XT.BIL_ChargeItemType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_67: {str(e)}")
# ============================================================================
# NODE node_68: $NDW_XT.BIL_ChargeItemType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing ChargeItemType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_68: $NDW_XT.BIL_ChargeItemType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_ChargeItemType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_ChargeItemType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_68 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ChargeItemType_Id_XT - Node node_68 ===")
    log_df_info(final_df_node_68, "Source DataFrame Information")
    
    logger.info("✓ Node node_68 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_68 ($NDW_XT.BIL_ChargeItemType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_68: {str(e)}")
# ============================================================================
# NODE node_69: $ndw_type.ChargeItemType (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Target table used for checking existing ChargeItemType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_69: $ndw_type.ChargeItemType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "ChargeItemType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read data from Unity Catalog
    final_df_node_69 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ChargeItemType - Node node_69 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_69.columns)}")
    final_df_node_69.printSchema()
    final_df_node_69.show(3, truncate=False)
    
    logger.info("✓ Node node_69 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_69 ($ndw_type.ChargeItemType): {str(e)}")
    raise Exception(f"Critical error in Node node_69: {str(e)}")
# ============================================================================
# NODE node_70: $WORKDB_BILLING.TW5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_70: $WORKDB_BILLING.TW5A0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5a0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5a0_incr.csv"  # Update path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_70 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5A0_INCR - Node node_70 ===")
    log_df_info(final_df_node_70, "Source DataFrame Information")
    
    logger.info("✓ Node node_70 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_70 ($WORKDB_BILLING.TW5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_70: {str(e)}")
# ============================================================================
# NODE node_71: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum CompanyType_Id and check for existing companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_71: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_CompanyType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_CompanyType_Id_XT.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_71 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_71 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_71.columns)}")
    final_df_node_71.printSchema()
    final_df_node_71.show(3, truncate=False)
    
    logger.info("✓ Node node_71 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_71 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_71: {str(e)}")
# ============================================================================
# NODE node_72: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_72: $WORKDB_BILLING.TW5TD_INCR (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5td_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5td_incr.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_72 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_72 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_72.columns)}")
    final_df_node_72.printSchema()
    final_df_node_72.show(3, truncate=False)
    
    logger.info("✓ Node node_72 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_72 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_72: {str(e)}")
# ============================================================================
# NODE node_73: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to check existing CompanyType_Id values and filter out existing companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_73: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_CompanyType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_CompanyType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_73 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_73 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_73.columns)}")
    final_df_node_73.printSchema()
    final_df_node_73.show(3, truncate=False)
    
    logger.info("✓ Node node_73 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_73 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_73: {str(e)}")
# ============================================================================
# NODE node_74: $WORKDB_BILLING.TW5Z0_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_74: $WORKDB_BILLING.TW5Z0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"  # Assuming default schema unless specified
    source_table = "TW5Z0_INCR"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5Z0_INCR.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_74 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5Z0_INCR - Node node_74 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_74.columns)}")
    final_df_node_74.printSchema()
    final_df_node_74.show(3, truncate=False)
    
    logger.info("✓ Node node_74 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_74 ($WORKDB_BILLING.TW5Z0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_74: {str(e)}")
# ============================================================================
# NODE node_75: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum CompanyType_Id and filter existing companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_75: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_companytype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_CompanyType_Id_XT.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_75 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_75 ===")
    log_df_info(final_df_node_75, "Source DataFrame Information")
    
    logger.info("✓ Node node_75 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_75 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_75: {str(e)}")
# ============================================================================
# NODE node_76: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_76: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_76 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_76 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_76.columns)}")
    final_df_node_76.printSchema()
    final_df_node_76.show(3, truncate=False)
    
    logger.info("✓ Node node_76 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_76 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_76: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv") \
                .option("header", "true") \
                .option("inferSchema", "true") \
                .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_77: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum CompanyType_Id and filter existing companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_77: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_companytype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_CompanyType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_77 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_77, "Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_77")
    
    logger.info("✓ Node node_77 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_77 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_77: {str(e)}")
# ============================================================================
# NODE node_78: $WORKDB_BILLING.TW6G0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_78: $WORKDB_BILLING.TW6G0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw6g0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw6g0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_78 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6G0_INCR - Node node_78 ===")
    log_df_info(final_df_node_78, "Source DataFrame Information")
    
    logger.info("✓ Node node_78 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_78 ($WORKDB_BILLING.TW6G0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_78: {str(e)}")
# ============================================================================
# NODE node_79: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum CompanyType_Id and check for existing companies.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_79: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_CompanyType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_CompanyType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_79 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_79 ===")
    log_df_info(final_df_node_79, "Source DataFrame Information")
    
    logger.info("✓ Node node_79 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_79 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_79: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_80: $NDW_XT.BIL_CompanyType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing company type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_80: $NDW_XT.BIL_CompanyType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_companytype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_CompanyType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_80 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_80, "Source: $NDW_XT.BIL_CompanyType_Id_XT - Node node_80")
    
    logger.info("✓ Node node_80 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_80 ($NDW_XT.BIL_CompanyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_80: {str(e)}")
# ============================================================================
# NODE node_81: $ndw_type.CompanyType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing CompanyType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_81: $ndw_type.CompanyType (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "CompanyType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_81 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.CompanyType - Node node_81 ===")
    log_df_info(final_df_node_81, "Source DataFrame Information")
    
    logger.info("✓ Node node_81 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_81 ($ndw_type.CompanyType): {str(e)}")
    raise Exception(f"Critical error in Node node_81: {str(e)}")
# ============================================================================
# NODE node_82: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental data for distribution channel codes and external systems.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_82: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$WORKDB_BILLING.TZ4A0_INCR"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_82 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_82 ===")
    log_df_info(final_df_node_82, "Source DataFrame Information")
    
    logger.info("✓ Node node_82 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_82 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_82: {str(e)}")
# ============================================================================
# NODE node_83: $NDW_XT.BIL_DisChannelType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum distribution channel type ID for calculation and to check existing distribution channel codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_83: $NDW_XT.BIL_DisChannelType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_DisChannelType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_DisChannelType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_83 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DisChannelType_Id_XT - Node node_83 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_83.columns)}")
    final_df_node_83.printSchema()
    final_df_node_83.show(3, truncate=False)
    
    logger.info("✓ Node node_83 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_83 ($NDW_XT.BIL_DisChannelType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_83: {str(e)}")
# ============================================================================
# NODE node_84: $WORKDB_BILLING.TZ5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for distribution channel codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_84: $WORKDB_BILLING.TZ5A0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tz5a0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tz5a0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_84 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ5A0_INCR - Node node_84 ===")
    log_df_info(final_df_node_84, "Source Data Loaded")
    
    logger.info("✓ Node node_84 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_84 ($WORKDB_BILLING.TZ5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_84: {str(e)}")
# ============================================================================
# NODE node_85: $NDW_XT.BIL_DisChannelType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to check existing distribution channel codes 
#              and retrieve the maximum distribution channel type ID.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_85: $NDW_XT.BIL_DisChannelType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DisChannelType_Id_XT"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_DisChannelType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_85 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DisChannelType_Id_XT - Node node_85 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_85.columns)}")
    final_df_node_85.printSchema()
    final_df_node_85.show(3, truncate=False)
    
    logger.info("✓ Node node_85 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_85 ($NDW_XT.BIL_DisChannelType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_85: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_86: $NDW_XT.BIL_DisChannelType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing distribution channel type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_86: $NDW_XT.BIL_DisChannelType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dischanneltype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_DisChannelType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_86 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_86, "Source Node node_86: $NDW_XT.BIL_DisChannelType_Id_XT")
    
    logger.info("✓ Node node_86 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_86 ($NDW_XT.BIL_DisChannelType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_86: {str(e)}")
# ============================================================================
# NODE node_87: $ndw_type.DistributionChannelType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing DistributionChannelType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_87: $ndw_type.DistributionChannelType")
    
    # Define the Unity Catalog table name
    table_name = "genai-powerbi_catalog.ndw_type.DistributionChannelType"
    local_path = "/mnt/data/ndw_type_DistributionChannelType.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper
    final_df_node_87 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.DistributionChannelType - Node node_87 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_87.columns)}")
    final_df_node_87.printSchema()
    final_df_node_87.show(3, truncate=False)
    
    logger.info("✓ Node node_87 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_87 ($ndw_type.DistributionChannelType): {str(e)}")
    raise Exception(f"Critical error in Node node_87: {str(e)}")
# ============================================================================
# NODE node_88: $WORKDB_BILLING.TW5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental Dunning_Code data to be processed.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_88: $WORKDB_BILLING.TW5A0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_88 = read_table(
        table_name=full_table_name, 
        local_file_path=None  # No fallback file path provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5A0_INCR - Node node_88 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_88.columns)}")
    final_df_node_88.printSchema()
    final_df_node_88.show(3, truncate=False)
    
    logger.info("✓ Node node_88 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_88 ($WORKDB_BILLING.TW5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_88: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_89: $NDW_XT.BIL_DunningCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum DunningCodeType_Id and check for existing Dunning_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_89: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_DunningCodeType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_89 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_89, "Source Node node_89: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    logger.info("✓ Node node_89 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_89 ($NDW_XT.BIL_DunningCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_89: {str(e)}")
# ============================================================================
# NODE node_90: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_90: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "default"
    table_name = "TW5TD_INCR"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_90 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_90 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_90.columns)}")
    final_df_node_90.printSchema()
    final_df_node_90.show(3, truncate=False)
    
    logger.info("✓ Node node_90 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_90 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_90: {str(e)}")
# ============================================================================
# NODE node_91: $NDW_XT.BIL_DunningCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum DunningCodeType_Id and validate existing Dunning_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_91: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_dunningcodetype_id_xt.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_91 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DunningCodeType_Id_XT - Node node_91 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_91.columns)}")
    final_df_node_91.printSchema()
    final_df_node_91.show(3, truncate=False)
    
    logger.info("✓ Node node_91 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_91 ($NDW_XT.BIL_DunningCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_91: {str(e)}")
# ============================================================================
# NODE node_92: $WORKDB_BILLING.TW5E0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental Dunning_Code data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_92: $WORKDB_BILLING.TW5E0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5e0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_92 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5E0_INCR - Node node_92 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_92.columns)}")
    final_df_node_92.printSchema()
    final_df_node_92.show(3, truncate=False)
    
    logger.info("✓ Node node_92 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_92 ($WORKDB_BILLING.TW5E0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_92: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_93: $NDW_XT.BIL_DunningCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum DunningCodeType_Id and filter existing Dunning_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_93: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_dunningcodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_93 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_93, "Source Node node_93: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    logger.info("✓ Node node_93 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_93 ($NDW_XT.BIL_DunningCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_93: {str(e)}")
# ============================================================================
# NODE node_94: $NDW_XT.BIL_DunningCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing Dunning Code Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_94: $NDW_XT.BIL_DunningCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_DunningCodeType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_DunningCodeType_Id_XT.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_94 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DunningCodeType_Id_XT - Node node_94 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_94.columns)}")
    final_df_node_94.printSchema()
    final_df_node_94.show(3, truncate=False)
    
    logger.info("✓ Node node_94 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_94 ($NDW_XT.BIL_DunningCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_94: {str(e)}")
# ============================================================================
# NODE node_95: $ndw_type.DunningCodeType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for comparison to ensure no duplicate DunningCodeType_Id values are inserted.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_95: $ndw_type.DunningCodeType (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "DunningCodeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_95 = read_table(
        table_name=table_name,
        local_file_path=None  # No fallback file path provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.DunningCodeType - Node node_95 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_95.columns)}")
    final_df_node_95.printSchema()
    final_df_node_95.show(3, truncate=False)
    
    logger.info("✓ Node node_95 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_95 ($ndw_type.DunningCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_95: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_96: $WORKDB_BILLING.TZ5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental EFT statement indicators.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_96: $WORKDB_BILLING.TZ5A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_96 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_96, "Source Node node_96: $WORKDB_BILLING.TZ5A0_INCR")
    
    logger.info("✓ Node node_96 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_96 ($WORKDB_BILLING.TZ5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_96: {str(e)}")
# ============================================================================
# NODE node_97: $NDW_XT.BIL_EFTStatementType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum EFTStatementType_Id and filter existing EFT_Statement_Ind values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_97: $NDW_XT.BIL_EFTStatementType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_EFTStatementType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_97 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_EFTStatementType_Id_Xt - Node node_97 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_97.columns)}")
    final_df_node_97.printSchema()
    final_df_node_97.show(3, truncate=False)
    
    logger.info("✓ Node node_97 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_97 ($NDW_XT.BIL_EFTStatementType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_97: {str(e)}")
# ============================================================================
# NODE node_98: $NDW_XT.BIL_EFTStatementType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing EFT statement type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_98: $NDW_XT.BIL_EFTStatementType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_EFTStatementType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_98 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_EFTStatementType_Id_Xt - Node node_98 ===")
    log_df_info(final_df_node_98, "Source DataFrame Information")
    
    logger.info("✓ Node node_98 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_98 ($NDW_XT.BIL_EFTStatementType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_98: {str(e)}")
# ============================================================================
# NODE node_99: $ndw_type.EFTStatementType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing EFT statement type IDs to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_99: $ndw_type.EFTStatementType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "EFTStatementType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_99 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.EFTStatementType - Node node_99 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_99.columns)}")
    final_df_node_99.printSchema()
    final_df_node_99.show(3, truncate=False)
    
    logger.info("✓ Node node_99 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_99 ($ndw_type.EFTStatementType): {str(e)}")
    raise Exception(f"Critical error in Node node_99: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_100: $WORKDB_BILLING.TU5P0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_100: $WORKDB_BILLING.TU5P0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TU5P0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TU5P0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_100 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_100, "Source Node node_100: $WORKDB_BILLING.TU5P0_INCR")
    
    logger.info("✓ Node node_100 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_100 ($WORKDB_BILLING.TU5P0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_100: {str(e)}")
# ============================================================================
# NODE node_101: $NDW_XT.BIL_EFTType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum ElectronicFundTransType_Id and filter existing Pay_Type values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_101: $NDW_XT.BIL_EFTType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_efttype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_EFTType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_101 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_EFTType_Id_XT - Node node_101 ===")
    log_df_info(final_df_node_101, "Source Data Loaded")
    
    logger.info("✓ Node node_101 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_101 ($NDW_XT.BIL_EFTType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_101: {str(e)}")
# ============================================================================
# NODE node_102: $NDW_XT.BIL_EFTType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing Electronic Fund Transfer Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_102: $NDW_XT.BIL_EFTType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_efttype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_102 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_EFTType_Id_XT - Node node_102 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_102.columns)}")
    final_df_node_102.printSchema()
    final_df_node_102.show(3, truncate=False)
    
    logger.info("✓ Node node_102 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_102 ($NDW_XT.BIL_EFTType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_102: {str(e)}")
# ============================================================================
# NODE node_103: $ndw_type.ElectronicFundTransType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used to check for existing ElectronicFundTransType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_103: $ndw_type.ElectronicFundTransType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "ElectronicFundTransType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/ElectronicFundTransType.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_103 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ElectronicFundTransType - Node node_103 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_103.columns)}")
    final_df_node_103.printSchema()
    final_df_node_103.show(3, truncate=False)
    
    logger.info("✓ Node node_103 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_103 ($ndw_type.ElectronicFundTransType): {str(e)}")
    raise Exception(f"Critical error in Node node_103: {str(e)}")
# ============================================================================
# NODE node_104: $WORKDB_BILLING.TW6G0_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_104: $WORKDB_BILLING.TW6G0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw6g0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_104 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6G0_INCR - Node node_104 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_104.columns)}")
    final_df_node_104.printSchema()
    final_df_node_104.show(3, truncate=False)
    
    logger.info("✓ Node node_104 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_104 ($WORKDB_BILLING.TW6G0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_104: {str(e)}")
# ============================================================================
# NODE node_105: $NDW_XT.BIL_GenLedgerAccountType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum GeneralLedgerAccountType_Id and check for existing DW_Account_Num values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_105: $NDW_XT.BIL_GenLedgerAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_genledgeraccounttype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_GenLedgerAccountType_Id_XT.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_105 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_GenLedgerAccountType_Id_XT - Node node_105 ===")
    log_df_info(final_df_node_105, "Source Node node_105")
    
    logger.info("✓ Node node_105 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_105 ($NDW_XT.BIL_GenLedgerAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_105: {str(e)}")
# ============================================================================
# NODE node_106: $NDW_XT.Bil_GenLedgerAccountType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing General Ledger Account Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_106: $NDW_XT.Bil_GenLedgerAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.Bil_GenLedgerAccountType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for local testing)
    local_file_path = "/mnt/data/$NDW_XT_Bil_GenLedgerAccountType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_106 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.Bil_GenLedgerAccountType_Id_XT - Node node_106 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_106.columns)}")
    final_df_node_106.printSchema()
    final_df_node_106.show(3, truncate=False)
    
    logger.info("✓ Node node_106 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_106 ($NDW_XT.Bil_GenLedgerAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_106: {str(e)}")
# ============================================================================
# NODE node_107: $ndw_type.GeneralLedgerAccountType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing GeneralLedgerAccountType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_107: $ndw_type.GeneralLedgerAccountType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "GeneralLedgerAccountType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_107 = read_table(
        table_name=table_name,
        local_file_path=None  # No fallback file path provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.GeneralLedgerAccountType - Node node_107 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_107.columns)}")
    final_df_node_107.printSchema()
    final_df_node_107.show(3, truncate=False)
    
    logger.info("✓ Node node_107 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_107 ($ndw_type.GeneralLedgerAccountType): {str(e)}")
    raise Exception(f"Critical error in Node node_107: {str(e)}")
# ============================================================================
# NODE node_108: $WORKDB_BILLING.TW6G0_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental data for Event_Code.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_108: $WORKDB_BILLING.TW6G0_INCR (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw6g0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw6g0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_108 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6G0_INCR - Node node_108 ===")
    log_df_info(final_df_node_108, "Source DataFrame Information")
    
    logger.info("✓ Node node_108 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_108 ($WORKDB_BILLING.TW6G0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_108: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_109: $NDW_XT.BIL_GenLedgerEventType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum GeneralLedgerEventType_Id for calculation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_109: $NDW_XT.BIL_GenLedgerEventType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_GenLedgerEventType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_109 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_109, "Source Node node_109: $NDW_XT.BIL_GenLedgerEventType_Id_XT")
    
    logger.info("✓ Node node_109 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_109 ($NDW_XT.BIL_GenLedgerEventType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_109: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_110: $NDW_XT.BIL_GenLedgerEventType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing General Ledger Event Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_110: $NDW_XT.BIL_GenLedgerEventType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerEventType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_110 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_110, "Source: $NDW_XT.BIL_GenLedgerEventType_Id_XT - Node node_110")
    
    logger.info("✓ Node node_110 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_110 ($NDW_XT.BIL_GenLedgerEventType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_110: {str(e)}")
# ============================================================================
# NODE node_111: $ndw_type.GeneralLedgerEventType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used to check for existing GeneralLedgerEventType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_111: $ndw_type.GeneralLedgerEventType (Source)")
    
    # Define the Unity Catalog table name
    table_name = "genai_powerbi_catalog.ndw_type.GeneralLedgerEventType"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/ndw_type/GeneralLedgerEventType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_111 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.GeneralLedgerEventType - Node node_111 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_111.columns)}")
    final_df_node_111.printSchema()
    final_df_node_111.show(3, truncate=False)
    
    logger.info("✓ Node node_111 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_111 ($ndw_type.GeneralLedgerEventType): {str(e)}")
    raise Exception(f"Critical error in Node node_111: {str(e)}")
# ============================================================================
# NODE node_112: $WORKDB_BILLING.TW6G0_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental data for general ledger source codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_112: $WORKDB_BILLING.TW6G0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tw6g0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TW6G0_INCR.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper
    final_df_node_112 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6G0_INCR - Node node_112 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_112.columns)}")
    final_df_node_112.printSchema()
    final_df_node_112.show(3, truncate=False)
    
    logger.info("✓ Node node_112 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_112 ($WORKDB_BILLING.TW6G0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_112: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_113: $NDW_XT.BIL_GenLedgerSourceType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to check existing general ledger source codes and retrieve the maximum GeneralLedgerSourceType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_113: $NDW_XT.BIL_GenLedgerSourceType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_genledgersourcetype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerSourceType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_113 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_113, "Source: $NDW_XT.BIL_GenLedgerSourceType_Id_XT - Node node_113")
    
    logger.info("✓ Node node_113 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_113 ($NDW_XT.BIL_GenLedgerSourceType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_113: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_114: $NDW_XT.BIL_GenLedgerSourceType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing General Ledger Source Type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_114: $NDW_XT.BIL_GenLedgerSourceType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_genledgersourcetype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerSourceType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_114 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_114, "Source Node node_114: $NDW_XT.BIL_GenLedgerSourceType_Id_XT")
    
    logger.info("✓ Node node_114 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_114 ($NDW_XT.BIL_GenLedgerSourceType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_114: {str(e)}")
# ============================================================================
# NODE node_115: $ndw_type.GeneralLedgerSourceType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing GeneralLedgerSourceType_Id values and for inserting new records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_115: $ndw_type.GeneralLedgerSourceType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "GeneralLedgerSourceType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/GeneralLedgerSourceType.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_115 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.GeneralLedgerSourceType - Node node_115 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_115.columns)}")
    final_df_node_115.printSchema()
    final_df_node_115.show(3, truncate=False)
    
    logger.info("✓ Node node_115 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_115 ($ndw_type.GeneralLedgerSourceType): {str(e)}")
    raise Exception(f"Critical error in Node node_115: {str(e)}")
# ============================================================================
# NODE node_116: $WORKDB_BILLING.TA205_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_116: $WORKDB_BILLING.TA205_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "ta205_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_ta205_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_116 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TA205_INCR - Node node_116 ===")
    log_df_info(final_df_node_116, "Source DataFrame Information")
    
    logger.info("✓ Node node_116 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_116 ($WORKDB_BILLING.TA205_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_116: {str(e)}")
# ============================================================================
# NODE node_117: $NDW_XT.BIL_InputMethodType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum InputMethodType_Id and filter existing Diary_Entry_Flag values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_117: $NDW_XT.BIL_InputMethodType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai_powerbi_catalog.$NDW_XT.BIL_InputMethodType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_InputMethodType_Id_XT.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_117 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_InputMethodType_Id_XT - Node node_117 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_117.columns)}")
    final_df_node_117.printSchema()
    final_df_node_117.show(3, truncate=False)
    
    logger.info("✓ Node node_117 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_117 ($NDW_XT.BIL_InputMethodType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_117: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_118: $NDW_XT.BIL_InputMethodType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing input method type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_118: $NDW_XT.BIL_InputMethodType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_inputmethodtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_InputMethodType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_118 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_118, "Source Node node_118: $NDW_XT.BIL_InputMethodType_Id_XT")
    
    logger.info("✓ Node node_118 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_118 ($NDW_XT.BIL_InputMethodType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_118: {str(e)}")
# ============================================================================
# NODE node_119: $ndw_type.InputMethodType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing InputMethodType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_119: $ndw_type.InputMethodType (Source)")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.InputMethodType"
    local_path = "/mnt/data/ndw_type/InputMethodType.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_119 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.InputMethodType - Node node_119 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_119.columns)}")
    final_df_node_119.printSchema()
    final_df_node_119.show(3, truncate=False)
    
    logger.info("✓ Node node_119 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_119 ($ndw_type.InputMethodType): {str(e)}")
    raise Exception(f"Critical error in Node node_119: {str(e)}")
# ============================================================================
# NODE node_120: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_120: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tz4a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Update path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_120 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_120 ===")
    log_df_info(final_df_node_120, "Source DataFrame Information")
    
    logger.info("✓ Node node_120 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_120 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_120: {str(e)}")
# ============================================================================
# NODE node_121: $NDW_XT.BIL_MarketChannelType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum MarketChannelType_Id and filter existing Mar_Channel_Code values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_121: $NDW_XT.BIL_MarketChannelType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai_powerbi_catalog.ndw_xt.bil_marketchanneltype_id_xt"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_121 = spark.table(source_table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_MarketChannelType_Id_XT - Node node_121 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_121.columns)}")
    final_df_node_121.printSchema()
    final_df_node_121.show(3, truncate=False)
    
    logger.info("✓ Node node_121 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_121 ($NDW_XT.BIL_MarketChannelType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_121: {str(e)}")
# ============================================================================
# NODE node_122: $NDW_XT.BIL_MarketChannelType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing MarketChannelType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_122: $NDW_XT.BIL_MarketChannelType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_MarketChannelType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_MarketChannelType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_122 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_MarketChannelType_Id_XT - Node node_122 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_122.columns)}")
    final_df_node_122.printSchema()
    final_df_node_122.show(3, truncate=False)
    
    logger.info("✓ Node node_122 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_122 ($NDW_XT.BIL_MarketChannelType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_122: {str(e)}")
# ============================================================================
# NODE node_123: $ndw_type.MarketChannelType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing MarketChannelType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_123: $ndw_type.MarketChannelType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "MarketChannelType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_123 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.MarketChannelType - Node node_123 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_123.columns)}")
    final_df_node_123.printSchema()
    final_df_node_123.show(3, truncate=False)
    
    logger.info("✓ Node node_123 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_123 ($ndw_type.MarketChannelType): {str(e)}")
    raise Exception(f"Critical error in Node node_123: {str(e)}")
# ============================================================================
# NODE node_124: $WORKDB_BILLING.TW6G0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for match key processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_124: $WORKDB_BILLING.TW6G0_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw6g0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw6g0_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_124 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW6G0_INCR - Node node_124 ===")
    log_df_info(final_df_node_124, "Source DataFrame Information")
    
    logger.info("✓ Node node_124 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_124 ($WORKDB_BILLING.TW6G0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_124: {str(e)}")
# ============================================================================
# NODE node_125: $NDW_XT.BIL_MatchKeyType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum MatchKeyType_Id and check for existing match keys.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_125: $NDW_XT.BIL_MatchKeyType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_MatchKeyType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/NDW_XT_BIL_MatchKeyType_Id_XT.csv"  # Update path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_125 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_MatchKeyType_Id_XT - Node node_125 ===")
    log_df_info(final_df_node_125, "Source DataFrame Information")
    
    logger.info("✓ Node node_125 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_125 ($NDW_XT.BIL_MatchKeyType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_125: {str(e)}")
# ============================================================================
# NODE node_126: $WORKDB_BILLING.TW5Z0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Plan_Status.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_126: $WORKDB_BILLING.TW5Z0_INCR (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5z0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5z0_incr.csv"  # Update path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_126 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5Z0_INCR - Node node_126 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_126.columns)}")
    final_df_node_126.printSchema()
    final_df_node_126.show(3, truncate=False)
    
    logger.info("✓ Node node_126 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_126 ($WORKDB_BILLING.TW5Z0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_126: {str(e)}")
# ============================================================================
# NODE node_127: $NDW_XT.BIL_PlanStatusType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to check existing Plan_Status values and retrieve the maximum PlanStatusType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_127: $NDW_XT.BIL_PlanStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_PlanStatusType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"  # Example path
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_127 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PlanStatusType_Id_XT - Node node_127 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_127.columns)}")
    final_df_node_127.printSchema()
    final_df_node_127.show(3, truncate=False)
    
    logger.info("✓ Node node_127 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_127 ($NDW_XT.BIL_PlanStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_127: {str(e)}")
# ============================================================================
# NODE node_128: $NDW_XT.BIL_PlanStatusType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing PlanStatusType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_128: $NDW_XT.BIL_PlanStatusType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_PlanStatusType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_128 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PlanStatusType_Id_XT - Node node_128 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_128.columns)}")
    final_df_node_128.printSchema()
    final_df_node_128.show(3, truncate=False)
    
    logger.info("✓ Node node_128 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_128 ($NDW_XT.BIL_PlanStatusType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_128: {str(e)}")
# ============================================================================
# NODE node_129: $ndw_type.PlanStatusType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing PlanStatusType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_129: $ndw_type.PlanStatusType")
    
    # Define the Unity Catalog table name
    table_name = "genai-powerbi_catalog.ndw_type.PlanStatusType"
    
    # Define a fallback local file path (if applicable, replace with actual path)
    local_file_path = "/mnt/data/ndw_type/PlanStatusType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_129 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.PlanStatusType - Node node_129 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_129.columns)}")
    final_df_node_129.printSchema()
    final_df_node_129.show(3, truncate=False)
    
    logger.info("✓ Node node_129 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_129 ($ndw_type.PlanStatusType): {str(e)}")
    raise Exception(f"Critical error in Node node_129: {str(e)}")
# ============================================================================
# NODE node_130: $WORKDB_BILLING.TZ5BO_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for producer categories.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_130: $WORKDB_BILLING.TZ5BO_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tz5bo_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tz5bo_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_130 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ5BO_INCR - Node node_130 ===")
    log_df_info(final_df_node_130, "Source Data Loaded")
    
    logger.info("✓ Node node_130 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_130 ($WORKDB_BILLING.TZ5BO_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_130: {str(e)}")
# ============================================================================
# NODE node_131: $NDW_XT.BIL_ProducerType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum ProducerType_Id for calculation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_131: $NDW_XT.BIL_ProducerType_Id_XT")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProducerType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_131 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ProducerType_Id_XT - Node node_131 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_131.columns)}")
    final_df_node_131.printSchema()
    final_df_node_131.show(3, truncate=False)
    
    logger.info("✓ Node node_131 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_131 ($NDW_XT.BIL_ProducerType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_131: {str(e)}")
# ============================================================================
# NODE node_132: $NDW_XT.BIL_ProducerType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing producer type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_132: $NDW_XT.BIL_ProducerType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_ProducerType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_132 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ProducerType_Id_XT - Node node_132 ===")
    log_df_info(final_df_node_132, "Source DataFrame Information")
    
    logger.info("✓ Node node_132 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_132 ($NDW_XT.BIL_ProducerType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_132: {str(e)}")
# ============================================================================
# NODE node_133: $ndw_type.ProducerType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing ProducerType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_133: $ndw_type.ProducerType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "ProducerType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_133 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ProducerType - Node node_133 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_133.columns)}")
    final_df_node_133.printSchema()
    final_df_node_133.show(3, truncate=False)
    
    logger.info("✓ Node node_133 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_133 ($ndw_type.ProducerType): {str(e)}")
    raise Exception(f"Critical error in Node node_133: {str(e)}")
# ============================================================================
# NODE node_134: $WORKDB_BILLING.TW5T2_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Return Reason Codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_134: $WORKDB_BILLING.TW5T2_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai-powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5t2_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5T2_INCR.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_134 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T2_INCR - Node node_134 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_134.columns)}")
    final_df_node_134.printSchema()
    final_df_node_134.show(3, truncate=False)
    
    logger.info("✓ Node node_134 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_134 ($WORKDB_BILLING.TW5T2_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_134: {str(e)}")
# ============================================================================
# NODE node_135: $NDW_XT.BIL_ReasonReturnType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to find the maximum ReasonReturnType_Id and filter existing Return Reason Codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_135: $NDW_XT.BIL_ReasonReturnType_Id_Xt")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_xt.bil_reasonreturntype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_ReasonReturnType_Id_Xt.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_135 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ReasonReturnType_Id_Xt - Node node_135 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_135.columns)}")
    final_df_node_135.printSchema()
    final_df_node_135.show(3, truncate=False)
    
    logger.info("✓ Node node_135 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_135 ($NDW_XT.BIL_ReasonReturnType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_135: {str(e)}")
# ============================================================================
# NODE node_136: $NDW_XT.BIL_ReasonReturnType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table containing reason return type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_136: $NDW_XT.BIL_ReasonReturnType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_ReasonReturnType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_ReasonReturnType_Id_Xt.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_136 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ReasonReturnType_Id_Xt - Node node_136 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_136.columns)}")
    final_df_node_136.printSchema()
    final_df_node_136.show(3, truncate=False)
    
    logger.info("✓ Node node_136 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_136 ($NDW_XT.BIL_ReasonReturnType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_136: {str(e)}")
# ============================================================================
# NODE node_137: $ndw_type.ReasonReturnType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing ReasonReturnType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_137: $ndw_type.ReasonReturnType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "ReasonReturnType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/ndw_type/ReasonReturnType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_137 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ReasonReturnType - Node node_137 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_137.columns)}")
    final_df_node_137.printSchema()
    final_df_node_137.show(3, truncate=False)
    
    logger.info("✓ Node node_137 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_137 ($ndw_type.ReasonReturnType): {str(e)}")
    raise Exception(f"Critical error in Node node_137: {str(e)}")
# ============================================================================
# NODE node_138: $WORKDB_BILLING.TW5T2_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Receipt_Type.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_138: $WORKDB_BILLING.TW5T2_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5t2_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5T2_INCR.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_138 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T2_INCR - Node node_138 ===")
    log_df_info(final_df_node_138, "Source Data Loaded")
    
    logger.info("✓ Node node_138 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_138 ($WORKDB_BILLING.TW5T2_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_138: {str(e)}")
# ============================================================================
# NODE node_139: $NDW_XT.BIL_ReceiptType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to check existing Receipt_Type values and calculate max ReceiptType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_139: $NDW_XT.BIL_ReceiptType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_ReceiptType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_ReceiptType_Id_Xt.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_139 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ReceiptType_Id_Xt - Node node_139 ===")
    log_df_info(final_df_node_139, "Source DataFrame Information")
    
    logger.info("✓ Node node_139 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_139 ($NDW_XT.BIL_ReceiptType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_139: {str(e)}")
# ============================================================================
# NODE node_140: $NDW_XT.BIL_ReceiptType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing receipt type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_140: $NDW_XT.BIL_ReceiptType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_ReceiptType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT.BIL_ReceiptType_Id_Xt.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_140 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ReceiptType_Id_Xt - Node node_140 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_140.columns)}")
    final_df_node_140.printSchema()
    final_df_node_140.show(3, truncate=False)
    
    logger.info("✓ Node node_140 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_140 ($NDW_XT.BIL_ReceiptType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_140: {str(e)}")
# ============================================================================
# NODE node_141: $ndw_type.ReceiptType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing ReceiptType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_141: $ndw_type.ReceiptType (Source)")
    
    # Define the Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.receipt_type"
    local_path = "/mnt/data/ndw_type_receipt_type.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_141 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ReceiptType - Node node_141 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_141.columns)}")
    final_df_node_141.printSchema()
    final_df_node_141.show(3, truncate=False)
    
    logger.info("✓ Node node_141 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_141 ($ndw_type.ReceiptType): {str(e)}")
    raise Exception(f"Critical error in Node node_141: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_142: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for residential market codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_142: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_142 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_142, "Source Node node_142: $WORKDB_BILLING.TZ4A0_INCR")
    
    logger.info("✓ Node node_142 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_142 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_142: {str(e)}")
# ============================================================================
# NODE node_143: $NDW_XT.BIL_ResMarketType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to find the maximum ResidentialMarketType_Id and filter existing market codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_143: $NDW_XT.BIL_ResMarketType_Id_XT")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ResMarketType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_ResMarketType_Id_XT.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_143 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ResMarketType_Id_XT - Node node_143 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_143.columns)}")
    final_df_node_143.printSchema()
    final_df_node_143.show(3, truncate=False)
    
    logger.info("✓ Node node_143 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_143 ($NDW_XT.BIL_ResMarketType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_143: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_144: $NDW_XT.BIL_ResMarketType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing residential market type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_144: $NDW_XT.BIL_ResMarketType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_resmarkettype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_ResMarketType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_144 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_144, "Source: $NDW_XT.BIL_ResMarketType_Id_XT - Node node_144")
    
    logger.info("✓ Node node_144 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_144 ($NDW_XT.BIL_ResMarketType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_144: {str(e)}")
# ============================================================================
# NODE node_145: $ndw_type.ResidentialMarketType (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Target table used for checking existing ResidentialMarketType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_145: $ndw_type.ResidentialMarketType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "ResidentialMarketType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_145 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.ResidentialMarketType - Node node_145 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_145.columns)}")
    final_df_node_145.printSchema()
    final_df_node_145.show(3, truncate=False)
    
    logger.info("✓ Node node_145 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_145 ($ndw_type.ResidentialMarketType): {str(e)}")
    raise Exception(f"Critical error in Node node_145: {str(e)}")
# ============================================================================
# NODE node_146: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental transaction data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_146: $WORKDB_BILLING.TW5TD_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "default"
    table_name = "$WORKDB_BILLING.TW5TD_INCR"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_146 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_146 ===")
    log_df_info(final_df_node_146, "Source DataFrame Information")
    
    logger.info("✓ Node node_146 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_146 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_146: {str(e)}")
# ============================================================================
# NODE node_147: $NDW_XT.BIL_TransactionType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to check existing transaction types and calculate max TransactionType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_147: $NDW_XT.BIL_TransactionType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_TransactionType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_TransactionType_Id_XT.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_147 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_TransactionType_Id_XT - Node node_147 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_147.columns)}")
    final_df_node_147.printSchema()
    final_df_node_147.show(3, truncate=False)
    
    logger.info("✓ Node node_147 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_147 ($NDW_XT.BIL_TransactionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_147: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_148: $NDW_XT.BIL_TransactionType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing transaction type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_148: $NDW_XT.BIL_TransactionType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_transactiontype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_TransactionType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_148 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_148, "Source Node node_148: $NDW_XT.BIL_TransactionType_Id_XT")
    
    logger.info("✓ Node node_148 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_148 ($NDW_XT.BIL_TransactionType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_148: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_149: $ndw_type.TransactionType (Source)
# Container: Batch 1 (Active)
# Description: Target table used for checking existing TransactionType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_149: $ndw_type.TransactionType")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_type.TransactionType"
    local_path = "/mnt/data/ndw_type/TransactionType.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_149 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_149, "Source Node node_149: $ndw_type.TransactionType")
    
    logger.info("✓ Node node_149 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_149 ($ndw_type.TransactionType): {str(e)}")
    raise Exception(f"Critical error in Node node_149: {str(e)}")
# ============================================================================
# NODE node_150: $WORKDB_BILLING.TW5A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for Account_Type.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_150: $WORKDB_BILLING.TW5A0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5a0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tw5a0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_150 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5A0_INCR - Node node_150 ===")
    log_df_info(final_df_node_150, "Source DataFrame Information")
    
    logger.info("✓ Node node_150 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_150 ($WORKDB_BILLING.TW5A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_150: {str(e)}")
# ============================================================================
# NODE node_151: $NDW_XT.BIL_BillingAccountType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to check existing Account_Type values and retrieve the maximum BillingAccountType_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_151: $NDW_XT.BIL_BillingAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BillingAccountType_Id_XT"
    
    # Define fallback local file path (if applicable, for local testing)
    local_file_path = "/mnt/data/$NDW_XT_BIL_BillingAccountType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_151 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BillingAccountType_Id_XT - Node node_151 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_151.columns)}")
    final_df_node_151.printSchema()
    final_df_node_151.show(3, truncate=False)
    
    logger.info("✓ Node node_151 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_151 ($NDW_XT.BIL_BillingAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_151: {str(e)}")
# ============================================================================
# NODE node_152: $NDW_XT.BIL_BillingAccountType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing billing account type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_152: $NDW_XT.BIL_BillingAccountType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_BillingAccountType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_BillingAccountType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_152 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BillingAccountType_Id_XT - Node node_152 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_152.columns)}")
    final_df_node_152.printSchema()
    final_df_node_152.show(3, truncate=False)
    
    logger.info("✓ Node node_152 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_152 ($NDW_XT.BIL_BillingAccountType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_152: {str(e)}")
# ============================================================================
# NODE node_153: $ndw_type.BillingAccountType (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Target table used for checking existing BillingAccountType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_153: $ndw_type.BillingAccountType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "BillingAccountType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_153 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.BillingAccountType - Node node_153 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_153.columns)}")
    final_df_node_153.printSchema()
    final_df_node_153.show(3, truncate=False)
    
    logger.info("✓ Node node_153 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_153 ($ndw_type.BillingAccountType): {str(e)}")
    raise Exception(f"Critical error in Node node_153: {str(e)}")
# ============================================================================
# NODE node_154: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for physical systems.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_154: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tz4a0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tz4a0_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_154 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_154 ===")
    log_df_info(final_df_node_154, "Source Node node_154")
    
    logger.info("✓ Node node_154 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_154 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_154: {str(e)}")
# ============================================================================
# NODE node_155: $NDW_XT.BIL_PolicyPhysicalSystem_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to check existing PolicyPhysicalSystem values and retrieve the maximum PolicyPhysicalSystem_Id.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_155: $NDW_XT.BIL_PolicyPhysicalSystem_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_PolicyPhysicalSystem_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/NDW_XT_BIL_PolicyPhysicalSystem_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_155 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PolicyPhysicalSystem_Id_XT - Node node_155 ===")
    log_df_info(final_df_node_155, "Source DataFrame Information")
    
    logger.info("✓ Node node_155 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_155 ($NDW_XT.BIL_PolicyPhysicalSystem_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_155: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_156: $NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table containing PolicyPhysicalSystem data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_156: $NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_PolicyPhysicalSystem_Id_Xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_156 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_156, "Source Node node_156: $NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt")
    
    logger.info("✓ Node node_156 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_156 ($NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_156: {str(e)}")
# ============================================================================
# NODE node_157: $ndw_type.PolicyPhysicalSystemType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing PolicyPhysicalSystemType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_157: $ndw_type.PolicyPhysicalSystemType (Source)")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.PolicyPhysicalSystemType"
    local_path = "/mnt/data/ndw_type/PolicyPhysicalSystemType.csv"  # Example fallback path
    
    # Read data using the universal read_table helper function
    final_df_node_157 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.PolicyPhysicalSystemType - Node node_157 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_157.columns)}")
    final_df_node_157.printSchema()
    final_df_node_157.show(3, truncate=False)
    
    logger.info("✓ Node node_157 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_157 ($ndw_type.PolicyPhysicalSystemType): {str(e)}")
    raise Exception(f"Critical error in Node node_157: {str(e)}")
# ============================================================================
# NODE node_158: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing PolicyBrand information.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_158: $WORKDB_BILLING.TZ4A0_INCR (Source)")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"  # Fallback path for local file
    
    # Read data using the universal read_table helper function
    final_df_node_158 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_158 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_158.columns)}")
    final_df_node_158.printSchema()
    final_df_node_158.show(3, truncate=False)
    
    logger.info("✓ Node node_158 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_158 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_158: {str(e)}")
# ============================================================================
# NODE node_159: $NDW_XT.BIL_PolicyBrand_Id_Xt (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve existing PolicyBrand_Id values for comparison and aggregation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_159: $NDW_XT.BIL_PolicyBrand_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"  # Assuming default schema unless specified
    source_table = "BIL_PolicyBrand_Id_Xt"
    
    # Full table name in Unity Catalog
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_159 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PolicyBrand_Id_Xt - Node node_159 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_159.columns)}")
    final_df_node_159.printSchema()
    final_df_node_159.show(3, truncate=False)
    
    logger.info("✓ Node node_159 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_159 ($NDW_XT.BIL_PolicyBrand_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_159: {str(e)}")
# ============================================================================
# NODE node_160: $NDW_XT.BIL_PolicyBrand_Id_Xt (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing PolicyBrand_Id and PolicyBrand data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_160: $NDW_XT.BIL_PolicyBrand_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"  # Assuming default schema unless specified
    source_table = "BIL_PolicyBrand_Id_Xt"
    
    # Full table name in Unity Catalog
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_160 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PolicyBrand_Id_Xt - Node node_160 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_160.columns)}")
    final_df_node_160.printSchema()
    final_df_node_160.show(3, truncate=False)
    
    logger.info("✓ Node node_160 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_160 ($NDW_XT.BIL_PolicyBrand_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_160: {str(e)}")
# ============================================================================
# NODE node_161: $ndw_type.PolicyBrandType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing PolicyBrandType_Id values and for inserting new records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_161: $ndw_type.PolicyBrandType")
    
    # Define the Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.PolicyBrandType"
    local_path = "/mnt/data/ndw_type/PolicyBrandType.csv"  # Fallback path for local file
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_161 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.PolicyBrandType - Node node_161 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_161.columns)}")
    final_df_node_161.printSchema()
    final_df_node_161.show(3, truncate=False)
    
    logger.info("✓ Node node_161 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_161 ($ndw_type.PolicyBrandType): {str(e)}")
    raise Exception(f"Critical error in Node node_161: {str(e)}")
# ============================================================================
# NODE node_162: $WORKDB_BILLING.TW5E0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for document types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_162: $WORKDB_BILLING.TW5E0_INCR")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai_powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5e0_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5E0_INCR.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_162 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5E0_INCR - Node node_162 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_162.columns)}")
    final_df_node_162.printSchema()
    final_df_node_162.show(3, truncate=False)
    
    logger.info("✓ Node node_162 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_162 ($WORKDB_BILLING.TW5E0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_162: {str(e)}")
# ============================================================================
# NODE node_163: $NDW_XT.BIL_DocumentType_ID_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to fetch the maximum DocumentType_ID and filter existing document types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_163: $NDW_XT.BIL_DocumentType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_163 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DocumentType_ID_XT - Node node_163 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_163.columns)}")
    final_df_node_163.printSchema()
    final_df_node_163.show(3, truncate=False)
    
    logger.info("✓ Node node_163 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_163 ($NDW_XT.BIL_DocumentType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_163: {str(e)}")
# ============================================================================
# NODE node_164: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_164: $WORKDB_BILLING.TW5TD_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai-powerbi_catalog"
    schema_name = "workdb_billing"
    table_name = "tw5td_incr"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5td_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_164 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_164 ===")
    log_df_info(final_df_node_164, "Source DataFrame Information")
    
    logger.info("✓ Node node_164 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_164 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_164: {str(e)}")
# ============================================================================
# NODE node_165: $NDW_XT.BIL_DocumentType_ID_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum DocumentType_ID and filter existing document types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_165: $NDW_XT.BIL_DocumentType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_165 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DocumentType_ID_XT - Node node_165 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_165.columns)}")
    final_df_node_165.printSchema()
    final_df_node_165.show(3, truncate=False)
    
    logger.info("✓ Node node_165 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_165 ($NDW_XT.BIL_DocumentType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_165: {str(e)}")
# ============================================================================
# NODE node_166: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_166: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5td_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_166 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_166 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_166.columns)}")
    final_df_node_166.printSchema()
    final_df_node_166.show(3, truncate=False)
    
    logger.info("✓ Node node_166 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_166 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_166: {str(e)}")
# ============================================================================
# NODE node_167: $NDW_XT.BIL_DocumentType_ID_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum DocumentType_ID and filter existing document types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_167: $NDW_XT.BIL_DocumentType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"  # Fallback path if catalog read fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_167 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DocumentType_ID_XT - Node node_167 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_167.columns)}")
    final_df_node_167.printSchema()
    final_df_node_167.show(3, truncate=False)
    
    logger.info("✓ Node node_167 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_167 ($NDW_XT.BIL_DocumentType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_167: {str(e)}")
# ============================================================================
# NODE node_168: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_168: $WORKDB_BILLING.TW5TD_INCR (Source)")
    
    # Define source table name in Unity Catalog
    catalog_name = "genai-powerbi_catalog"
    schema_name = "default"
    table_name = "$WORKDB_BILLING.TW5TD_INCR"
    full_table_name = f"{catalog_name}.{schema_name}.{table_name}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_168 = read_table(full_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_168 ===")
    log_df_info(final_df_node_168, "Source Node node_168")
    
    logger.info("✓ Node node_168 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_168 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_168: {str(e)}")
# ============================================================================
# NODE node_169: $NDW_XT.BIL_DocumentType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum DocumentType_ID and filter existing document types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_169: $NDW_XT.BIL_DocumentType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_DocumentType_ID_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_DocumentType_ID_XT.csv"  # Update path if necessary
    
    # Read data using the universal read_table helper
    final_df_node_169 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DocumentType_ID_XT - Node node_169 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_169.columns)}")
    final_df_node_169.printSchema()
    final_df_node_169.show(3, truncate=False)
    
    logger.info("✓ Node node_169 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_169 ($NDW_XT.BIL_DocumentType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_169: {str(e)}")
# ============================================================================
# NODE node_170: $NDW_XT.BIL_DocumentType_ID_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing document type information to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_170: $NDW_XT.BIL_DocumentType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_DocumentType_ID_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_170 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_DocumentType_ID_XT - Node node_170 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_170.columns)}")
    final_df_node_170.printSchema()
    final_df_node_170.show(3, truncate=False)
    
    logger.info("✓ Node node_170 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_170 ($NDW_XT.BIL_DocumentType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_170: {str(e)}")
# ============================================================================
# NODE node_171: $ndw_type.DocumentType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used to check for existing DocumentType_IDs and to insert new records.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_171: $ndw_type.DocumentType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "DocumentType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for local testing)
    local_file_path = "/mnt/data/ndw_type/DocumentType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_171 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.DocumentType - Node node_171 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_171.columns)}")
    final_df_node_171.printSchema()
    final_df_node_171.show(3, truncate=False)
    
    logger.info("✓ Node node_171 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_171 ($ndw_type.DocumentType): {str(e)}")
    raise Exception(f"Critical error in Node node_171: {str(e)}")
# ============================================================================
# NODE node_172: $WORKDB_BILLING.TW5E0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for OWING_PLAN_CODE.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_172: $WORKDB_BILLING.TW5E0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5e0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_172 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5E0_INCR - Node node_172 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_172.columns)}")
    final_df_node_172.printSchema()
    final_df_node_172.show(3, truncate=False)
    
    logger.info("✓ Node node_172 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_172 ($WORKDB_BILLING.TW5E0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_172: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_173: $NDW_XT.BIL_PlanCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum PlanCodeType_Id and filter existing PlanCodeType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_173: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_plancodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_173 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_173, "Source Node node_173: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    logger.info("✓ Node node_173 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_173 ($NDW_XT.BIL_PlanCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_173: {str(e)}")
# ============================================================================
# NODE node_174: $WORKDB_BILLING.TW5E0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_174: $WORKDB_BILLING.TW5E0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5e0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5e0_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_174 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5E0_INCR - Node node_174 ===")
    log_df_info(final_df_node_174, "Source DataFrame Information")
    
    logger.info("✓ Node node_174 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_174 ($WORKDB_BILLING.TW5E0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_174: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_175: $NDW_XT.BIL_PlanCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum PlanCodeType_Id and filter existing PlanCodeType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_175: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_plancodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_175 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_175, "Source Node node_175: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    logger.info("✓ Node node_175 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_175 ($NDW_XT.BIL_PlanCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_175: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_176: $WORKDB_BILLING.TW5E0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_176: $WORKDB_BILLING.TW5E0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5E0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5E0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_176 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_176, "Source Node node_176: $WORKDB_BILLING.TW5E0_INCR")
    
    logger.info("✓ Node node_176 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_176 ($WORKDB_BILLING.TW5E0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_176: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_177: $NDW_XT.BIL_PlanCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum PlanCodeType_Id and filter existing PlanCodeType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_177: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_plancodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_177 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_177, "Source Node node_177: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    logger.info("✓ Node node_177 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_177 ($NDW_XT.BIL_PlanCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_177: {str(e)}")
# ============================================================================
# NODE node_178: $WORKDB_BILLING.TW5Z0_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental data for PLAN_CODE.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_178: $WORKDB_BILLING.TW5Z0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5z0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_178 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5Z0_INCR - Node node_178 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_178.columns)}")
    final_df_node_178.printSchema()
    final_df_node_178.show(3, truncate=False)
    
    logger.info("✓ Node node_178 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_178 ($WORKDB_BILLING.TW5Z0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_178: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_179: $NDW_XT.BIL_PlanCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum PlanCodeType_Id and check for existing PlanCodeType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_179: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_plancodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_179 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_179, "Source Node node_179: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    logger.info("✓ Node node_179 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_179 ($NDW_XT.BIL_PlanCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_179: {str(e)}")
# ============================================================================
# NODE node_180: $NDW_XT.BIL_PlanCodeType_ID_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing PlanCodeType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_180: $NDW_XT.BIL_PlanCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_PlanCodeType_ID_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_PlanCodeType_ID_XT.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_180 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PlanCodeType_ID_XT - Node node_180 ===")
    log_df_info(final_df_node_180, "Source DataFrame Information")
    
    logger.info("✓ Node node_180 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_180 ($NDW_XT.BIL_PlanCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_180: {str(e)}")
# ============================================================================
# NODE node_181: $ndw_type.PlanCodeType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing PlanCodeType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_181: $ndw_type.PlanCodeType")
    
    # Define the Unity Catalog table name
    table_name = "genai-powerbi_catalog.ndw_type.PlanCodeType"
    
    # Define a fallback local file path (if applicable, for local testing)
    local_file_path = "/mnt/data/ndw_type/PlanCodeType.csv"
    
    # Read the data from Unity Catalog or fallback to local file
    final_df_node_181 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.PlanCodeType - Node node_181 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_181.columns)}")
    final_df_node_181.printSchema()
    final_df_node_181.show(3, truncate=False)
    
    logger.info("✓ Node node_181 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_181 ($ndw_type.PlanCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_181: {str(e)}")
# ============================================================================
# NODE node_182: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing incremental transaction data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_182: $WORKDB_BILLING.TW5TD_INCR (Source)")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5td_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_182 = read_table(table_name, local_file_path=None)  # No local file fallback provided
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5TD_INCR - Node node_182 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_182.columns)}")
    final_df_node_182.printSchema()
    final_df_node_182.show(3, truncate=False)
    
    logger.info("✓ Node node_182 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_182 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_182: {str(e)}")
# ============================================================================
# NODE node_183: $NDW_XT.BIL_TransactionGroupType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to retrieve the maximum TransactionGroupType_Id and check for existing TransactionGroupType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_183: $NDW_XT.BIL_TransactionGroupType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_TransactionGroupType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_TransactionGroupType_Id_XT.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_183 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_TransactionGroupType_Id_XT - Node node_183 ===")
    log_df_info(final_df_node_183, "Source DataFrame Information")
    
    logger.info("✓ Node node_183 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_183 ($NDW_XT.BIL_TransactionGroupType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_183: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_184: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental transaction data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_184: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_184 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_184, "Source Node node_184: $WORKDB_BILLING.TW5TD_INCR")
    
    logger.info("✓ Node node_184 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_184 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_184: {str(e)}")
# ============================================================================
# NODE node_185: $NDW_XT.BIL_TransactionGroupType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum TransactionGroupType_Id and validate TransactionGroupType_Cd.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_185: $NDW_XT.BIL_TransactionGroupType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_TransactionGroupType_Id_XT"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_TransactionGroupType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_185 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_TransactionGroupType_Id_XT - Node node_185 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_185.columns)}")
    final_df_node_185.printSchema()
    final_df_node_185.show(3, truncate=False)
    
    logger.info("✓ Node node_185 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_185 ($NDW_XT.BIL_TransactionGroupType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_185: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_186: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental transaction data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_186: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_186 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_186, "Source Node node_186: $WORKDB_BILLING.TW5TD_INCR")
    
    logger.info("✓ Node node_186 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_186 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_186: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_187: $NDW_XT.BIL_TransactionGroupType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum TransactionGroupType_Id and validate TransactionGroupType_Cd.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_187: $NDW_XT.BIL_TransactionGroupType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_transactiongrouptype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_TransactionGroupType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_187 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_187, "Source Node node_187: $NDW_XT.BIL_TransactionGroupType_Id_XT")
    
    logger.info("✓ Node node_187 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_187 ($NDW_XT.BIL_TransactionGroupType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_187: {str(e)}")
# ============================================================================
# NODE node_188: $NDW_XT.BIL_TransactionGroupType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing transaction group type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_188: $NDW_XT.BIL_TransactionGroupType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_TransactionGroupType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_TransactionGroupType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_188 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_TransactionGroupType_Id_XT - Node node_188 ===")
    log_df_info(final_df_node_188, "Source DataFrame Information")
    
    logger.info("✓ Node node_188 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_188 ($NDW_XT.BIL_TransactionGroupType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_188: {str(e)}")
# ============================================================================
# NODE node_189: $ndw_type.TransactionGroupType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing TransactionGroupType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_189: $ndw_type.TransactionGroupType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "TransactionGroupType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/TransactionGroupType.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_189 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.TransactionGroupType - Node node_189 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_189.columns)}")
    final_df_node_189.printSchema()
    final_df_node_189.show(3, truncate=False)
    
    logger.info("✓ Node node_189 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_189 ($ndw_type.TransactionGroupType): {str(e)}")
    raise Exception(f"Critical error in Node node_189: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================
# NODE node_190: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for subsystem notifications.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_190: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_190 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_190, "Source Node node_190: $WORKDB_BILLING.TW5TD_INCR")
    
    logger.info("✓ Node node_190 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_190 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_190: {str(e)}")
# ============================================================================
# NODE node_191: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table used to determine the maximum SubsystemNotifyCodeType_Id and filter existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_191: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_SubsystemNtfyCdType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_SubsystemNtfyCdType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_191 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT - Node node_191 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_191.columns)}")
    final_df_node_191.printSchema()
    final_df_node_191.show(3, truncate=False)
    
    logger.info("✓ Node node_191 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_191 ($NDW_XT.BIL_SubsystemNtfyCdType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_191: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_192: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_192: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tw5td_incr"
    local_path = "/mnt/data/workdb_billing_tw5td_incr.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_192 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_192, "Source Node node_192: $WORKDB_BILLING.TW5TD_INCR")
    
    logger.info("✓ Node node_192 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_192 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_192: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_193: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum SubsystemNotifyCodeType_Id and validate existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_193: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_subsystemntfycdtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_SubsystemNtfyCdType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback file
    final_df_node_193 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_193, "Source Node node_193: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    logger.info("✓ Node node_193 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_193 ($NDW_XT.BIL_SubsystemNtfyCdType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_193: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================

# ============================================================================
# NODE node_194: $WORKDB_BILLING.TW5TD_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_194: $WORKDB_BILLING.TW5TD_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_194 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_194, "Source Node node_194: $WORKDB_BILLING.TW5TD_INCR")
    
    logger.info("✓ Node node_194 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_194 ($WORKDB_BILLING.TW5TD_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_194: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_195: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum SubsystemNotifyCodeType_Id and filter existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_195: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_subsystemntfycdtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_SubsystemNtfyCdType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_195 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_195, "Source Node node_195: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    logger.info("✓ Node node_195 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_195 ($NDW_XT.BIL_SubsystemNtfyCdType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_195: {str(e)}")
# ============================================================================
# NODE node_196: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing subsystem notification code types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_196: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_SubsystemNtfyCdType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_SubsystemNtfyCdType_Id_XT.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_196 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_SubsystemNtfyCdType_Id_XT - Node node_196 ===")
    log_df_info(final_df_node_196, "Source DataFrame Information")
    
    logger.info("✓ Node node_196 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_196 ($NDW_XT.BIL_SubsystemNtfyCdType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_196: {str(e)}")
# ============================================================================
# NODE node_197: $ndw_type.SubsystemNotifyCodeType (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Target table used for checking existing subsystem notification code types.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_197: $ndw_type.SubsystemNotifyCodeType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "SubsystemNotifyCodeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_197 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.SubsystemNotifyCodeType - Node node_197 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_197.columns)}")
    final_df_node_197.printSchema()
    final_df_node_197.show(3, truncate=False)
    
    logger.info("✓ Node node_197 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_197 ($ndw_type.SubsystemNotifyCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_197: {str(e)}")
# ============================================================================
# NODE node_198: $WORKDB_BILLING.TW5T2_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental payment intent data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_198: $WORKDB_BILLING.TW5T2_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tw5t2_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing_tw5t2_incr.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_198 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5T2_INCR - Node node_198 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_198.columns)}")
    final_df_node_198.printSchema()
    final_df_node_198.show(3, truncate=False)
    
    logger.info("✓ Node node_198 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_198 ($WORKDB_BILLING.TW5T2_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_198: {str(e)}")
# ============================================================================
# NODE node_199: $NDW_XT.BIL_PymntIntentCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum PaymentIntentCodeType_Id and validate existing PaymentIntent_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_199: $NDW_XT.BIL_PymntIntentCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_pymntintentcodetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_PymntIntentCodeType_ID_XT.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_199 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PymntIntentCodeType_ID_XT - Node node_199 ===")
    log_df_info(final_df_node_199, "Source DataFrame Information")
    
    logger.info("✓ Node node_199 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_199 ($NDW_XT.BIL_PymntIntentCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_199: {str(e)}")
# ============================================================================
# NODE node_200: $NDW_XT.BIL_PymntIntentCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing payment intent code type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_200: $NDW_XT.BIL_PymntIntentCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_pymntintentcodetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_PymntIntentCodeType_ID_XT.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_200 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_PymntIntentCodeType_ID_XT - Node node_200 ===")
    log_df_info(final_df_node_200, "Source DataFrame")
    
    logger.info("✓ Node node_200 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_200 ($NDW_XT.BIL_PymntIntentCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_200: {str(e)}")
# ============================================================================
# NODE node_201: $ndw_type.PaymentIntentCodeType (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Target table used for checking existing records and inserting new ones.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_201: $ndw_type.PaymentIntentCodeType")
    
    # Define Unity Catalog table name and fallback local file path
    table_name = "genai-powerbi_catalog.ndw_type.PaymentIntentCodeType"
    local_path = "/mnt/data/ndw_type_PaymentIntentCodeType.csv"  # Example fallback path
    
    # Read data using the universal read_table helper function
    final_df_node_201 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.PaymentIntentCodeType - Node node_201 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_201.columns)}")
    final_df_node_201.printSchema()
    final_df_node_201.show(3, truncate=False)
    
    logger.info("✓ Node node_201 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_201 ($ndw_type.PaymentIntentCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_201: {str(e)}")
# ============================================================================
# NODE node_202: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_202: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"  # Fallback path if catalog read fails
    
    # Read data using the universal read_table helper function
    final_df_node_202 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A0_INCR - Node node_202 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_202.columns)}")
    final_df_node_202.printSchema()
    final_df_node_202.show(3, truncate=False)
    
    logger.info("✓ Node node_202 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_202 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_202: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_203: $NDW_XT.BIL_NonRenewalCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum NonRenewalCodeType_Id and validate against existing NonRenewalCodeType_Cd.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_203: $NDW_XT.BIL_NonRenewalCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.ndw_xt.bil_nonrenewalcodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_nonrenewalcodetype_id_xt.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_203 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_203, "Source Node node_203: $NDW_XT.BIL_NonRenewalCodeType_Id_XT")
    
    logger.info("✓ Node node_203 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_203 ($NDW_XT.BIL_NonRenewalCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_203: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_204: $WORKDB_BILLING.TZ4A0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_204: $WORKDB_BILLING.TZ4A0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the universal read_table helper
    final_df_node_204 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_204, "Source Node node_204: $WORKDB_BILLING.TZ4A0_INCR")
    
    logger.info("✓ Node node_204 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_204 ($WORKDB_BILLING.TZ4A0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_204: {str(e)}")
# ============================================================================
# NODE node_205: $NDW_XT.BIL_NonRenewalCodeType_Id_XT (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Source table used to fetch the maximum NonRenewalCodeType_Id and validate existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_205: $NDW_XT.BIL_NonRenewalCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_nonrenewalcodetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_205 = read_table(
        table_name=table_name,
        local_file_path=None  # No local file fallback provided for this source
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_NonRenewalCodeType_Id_XT - Node node_205 ===")
    log_df_info(final_df_node_205, step_name="Source Data Loaded")
    
    logger.info("✓ Node node_205 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_205 ($NDW_XT.BIL_NonRenewalCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_205: {str(e)}")
# ============================================================================
# NODE node_206: $NDW_XT.BIL_NonRenewalCodeType_Id_XT (Source)
# Container: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (Active)
# Description: Source table containing NonRenewalCodeType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_206: $NDW_XT.BIL_NonRenewalCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_NonRenewalCodeType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_NonRenewalCodeType_Id_XT.csv"  # Adjust path if needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_206 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_NonRenewalCodeType_Id_XT - Node node_206 ===")
    log_df_info(final_df_node_206, "Source DataFrame Information")
    
    logger.info("✓ Node node_206 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_206 ($NDW_XT.BIL_NonRenewalCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_206: {str(e)}")
# ============================================================================
# NODE node_207: $ndw_type.NonRenewalCodeType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing NonRenewalCodeType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_207: $ndw_type.NonRenewalCodeType")
    
    # Define the Unity Catalog table name
    table_name = "genai-powerbi_catalog.ndw_type.NonRenewalCodeType"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/ndw_type/NonRenewalCodeType.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_207 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.NonRenewalCodeType - Node node_207 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_207.columns)}")
    final_df_node_207.printSchema()
    final_df_node_207.show(3, truncate=False)
    
    logger.info("✓ Node node_207 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_207 ($ndw_type.NonRenewalCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_207: {str(e)}")
# ============================================================================
# NODE node_208: $WORKDB_BILLING.TZ4A1_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental billing data.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_208: $WORKDB_BILLING.TZ4A1_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tz4a1_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tz4a1_incr.csv"  # Adjust path as needed
    
    # Read data using the universal read_table helper
    final_df_node_208 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4A1_INCR - Node node_208 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_208.columns)}")
    final_df_node_208.printSchema()
    final_df_node_208.show(3, truncate=False)
    
    logger.info("✓ Node node_208 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_208 ($WORKDB_BILLING.TZ4A1_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_208: {str(e)}")
# ============================================================================
# NODE node_209: $NDW_XT.BIL_BillRcpntCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table used to retrieve the maximum BillRecipientCodeType_Id and validate BILL_TO values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_209: $NDW_XT.BIL_BillRcpntCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_billrcpntcodetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_BillRcpntCodeType_ID_XT.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_209 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_BillRcpntCodeType_ID_XT - Node node_209 ===")
    log_df_info(final_df_node_209, "Source DataFrame Information")
    
    logger.info("✓ Node node_209 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_209 ($NDW_XT.BIL_BillRcpntCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_209: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_210: $NDW_XT.BIL_BillRcpntCodeType_ID_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing BillRecipientCodeType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_210: $NDW_XT.BIL_BillRcpntCodeType_ID_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_billrcpntcodetype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_BillRcpntCodeType_ID_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_210 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_210, "Source Node node_210: $NDW_XT.BIL_BillRcpntCodeType_ID_XT")
    
    logger.info("✓ Node node_210 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_210 ($NDW_XT.BIL_BillRcpntCodeType_ID_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_210: {str(e)}")
# ============================================================================
# NODE node_211: $ndw_type.BillRecipientCodeType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing BillRecipientCodeType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_211: $ndw_type.BillRecipientCodeType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "BillRecipientCodeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_211 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.BillRecipientCodeType - Node node_211 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_211.columns)}")
    final_df_node_211.printSchema()
    final_df_node_211.show(3, truncate=False)
    
    logger.info("✓ Node node_211 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_211 ($ndw_type.BillRecipientCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_211: {str(e)}")
# ============================================================================
# NODE node_212: $WORKDB_BILLING.TZ4T0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_212: $WORKDB_BILLING.TZ4T0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tz4t0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/workdb_billing/tz4t0_incr.csv"  # Adjust path as needed
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_212 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4T0_INCR - Node node_212 ===")
    log_df_info(final_df_node_212, "Source Data Loaded")
    
    logger.info("✓ Node node_212 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_212 ($WORKDB_BILLING.TZ4T0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_212: {str(e)}")
# ============================================================================
# NODE node_213: $NDW_XT.BIL_TransReasonCodeType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to fetch the maximum TransactionReasonCodeType_Id and validate existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_213: $NDW_XT.BIL_TransReasonCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_transreasoncodetype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_TransReasonCodeType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_213 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_TransReasonCodeType_Id_XT - Node node_213 ===")
    log_df_info(final_df_node_213, "Source DataFrame Information")
    
    logger.info("✓ Node node_213 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_213 ($NDW_XT.BIL_TransReasonCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_213: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_214: $NDW_XT.BIL_TransReasonCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing transaction reason codes to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_214: $NDW_XT.BIL_TransReasonCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.ndw_xt.bil_transreasoncodetype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_TransReasonCodeType_Id_XT.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_214 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_214, "Source: $NDW_XT.BIL_TransReasonCodeType_Id_XT - Node node_214")
    
    logger.info("✓ Node node_214 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_214 ($NDW_XT.BIL_TransReasonCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_214: {str(e)}")
# ============================================================================
# NODE node_215: $ndw_type.TransactionReasonCodeType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing transaction reason codes and inserting new ones.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_215: $ndw_type.TransactionReasonCodeType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "TransactionReasonCodeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/TransactionReasonCodeType.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_215 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.TransactionReasonCodeType - Node node_215 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_215.columns)}")
    final_df_node_215.printSchema()
    final_df_node_215.show(3, truncate=False)
    
    logger.info("✓ Node node_215 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_215 ($ndw_type.TransactionReasonCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_215: {str(e)}")
# ============================================================================
# NODE node_216: $WORKDB_BILLING.TZ4H0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental history codes to be processed.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_216: $WORKDB_BILLING.TZ4H0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "workdb_billing"
    source_table = "tz4h0_incr"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Attempt to read the source table from Unity Catalog
    final_df_node_216 = spark.table(table_name)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4H0_INCR - Node node_216 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_216.columns)}")
    final_df_node_216.printSchema()
    final_df_node_216.show(3, truncate=False)
    
    logger.info("✓ Node node_216 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_216 ($WORKDB_BILLING.TZ4H0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_216: {str(e)}")
# ============================================================================
# NODE node_217: $NDW_XT.BIL_HistoryCodeType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to determine the maximum HistoryCodeType_Id and filter existing codes.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_217: $NDW_XT.BIL_HistoryCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_HistoryCodeType_Id_XT"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_HistoryCodeType_Id_XT.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_217 = read_table(source_table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_HistoryCodeType_Id_XT - Node node_217 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_217.columns)}")
    final_df_node_217.printSchema()
    final_df_node_217.show(3, truncate=False)
    
    logger.info("✓ Node node_217 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_217 ($NDW_XT.BIL_HistoryCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_217: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATION
# ============================================================================
# ============================================================================
# NODE node_218: $WORKDB_BILLING.TZ5H0_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental HISTORY_CODE data to be processed.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_218: $WORKDB_BILLING.TZ5H0_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TZ5H0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5H0_INCR.csv"  # Fallback path if catalog fails
    
    # Read data from Unity Catalog or fallback local file
    final_df_node_218 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_218, "Source Node node_218: $WORKDB_BILLING.TZ5H0_INCR")
    
    logger.info("✓ Node node_218 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_218 ($WORKDB_BILLING.TZ5H0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_218: {str(e)}")
# ============================================================================
# NODE node_219: $NDW_XT.BIL_HistoryCodeType_Id_XT (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table containing existing HistoryCodeType_Id and HistoryCodeType_Cd data for comparison and aggregation.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_219: $NDW_XT.BIL_HistoryCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$NDW_XT.BIL_HistoryCodeType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_HistoryCodeType_Id_XT.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_219 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_HistoryCodeType_Id_XT - Node node_219 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_219.columns)}")
    final_df_node_219.printSchema()
    final_df_node_219.show(3, truncate=False)
    
    logger.info("✓ Node node_219 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_219 ($NDW_XT.BIL_HistoryCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_219: {str(e)}")
# ============================================================================
# NODE node_220: $NDW_XT.BIL_HistoryCodeType_Id_XT (Source)
# Container: Batch 1 (Active)
# Description: Source table containing HistoryCodeType data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_220: $NDW_XT.BIL_HistoryCodeType_Id_XT")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_HistoryCodeType_Id_XT"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_HistoryCodeType_Id_XT.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_220 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_HistoryCodeType_Id_XT - Node node_220 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_220.columns)}")
    final_df_node_220.printSchema()
    final_df_node_220.show(3, truncate=False)
    
    logger.info("✓ Node node_220 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_220 ($NDW_XT.BIL_HistoryCodeType_Id_XT): {str(e)}")
    raise Exception(f"Critical error in Node node_220: {str(e)}")
# ============================================================================
# NODE node_221: $ndw_type.HistoryCodeType (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Target table used for checking existing HistoryCodeType_Id values to avoid duplicates.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_221: $ndw_type.HistoryCodeType")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_type"
    source_table = "HistoryCodeType"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_type/HistoryCodeType.csv"  # Update path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_221 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $ndw_type.HistoryCodeType - Node node_221 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_221.columns)}")
    final_df_node_221.printSchema()
    final_df_node_221.show(3, truncate=False)
    
    logger.info("✓ Node node_221 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_221 ($ndw_type.HistoryCodeType): {str(e)}")
    raise Exception(f"Critical error in Node node_221: {str(e)}")
# ============================================================================
# NODE node_222: $WORKDB_BILLING.TZ4P0_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental data for external systems.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_222: $WORKDB_BILLING.TZ4P0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$WORKDB_BILLING_TZ4P0_INCR"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TZ4P0_INCR.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_222 = read_table(
        table_name=f"{source_catalog}.{source_schema}.{source_table}",
        local_file_path=local_file_path
    )
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TZ4P0_INCR - Node node_222 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_222.columns)}")
    final_df_node_222.printSchema()
    final_df_node_222.show(3, truncate=False)
    
    logger.info("✓ Node node_222 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_222 ($WORKDB_BILLING.TZ4P0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_222: {str(e)}")
# ============================================================================
# NODE node_223: $NDW_XT.BIL_ExternalSystemType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum ExternalSystemType_Id 
#              and filter existing ExternalSystemType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_223: $NDW_XT.BIL_ExternalSystemType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "NDW_XT"
    source_table = "BIL_ExternalSystemType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"  # Example path
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_223 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ExternalSystemType_Id_Xt - Node node_223 ===")
    log_df_info(final_df_node_223, "Source DataFrame Information")
    
    logger.info("✓ Node node_223 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_223 ($NDW_XT.BIL_ExternalSystemType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_223: {str(e)}")
# ============================================================================
# IMPORTS AND SETUP
# ============================================================================
from pyspark.sql import SparkSession, functions as F, Row
from pyspark.sql.types import *
import logging

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Spark session is pre-initialized in Databricks
# Available as: spark

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def log_df_info(df, step_name):
    """
    Log DataFrame information safely (without expensive operations)
    NEVER use df.count() - it causes "Connection reset" errors
    """
    logger.info(f"=== {step_name} ===")
    logger.info(f"   └─ Columns: {len(df.columns)}")
    df.printSchema()
    try:
        logger.info(f"   └─ Sample (first 3 rows):")
        df.show(3, truncate=False)
    except Exception as e:
        logger.info(f"   └─ Could not show data: {str(e)}")

def read_table(table_name, local_file_path):
    """
    Universal data reader - tries Unity Catalog first, then local file
    Args:
        table_name: Unity Catalog table name (catalog.schema.table)
        local_file_path: Fallback local CSV path
    Returns:
        Spark DataFrame
    """
    try:
        logger.info(f"Attempting to read from Unity Catalog: {table_name}")
        df = spark.table(table_name)
        logger.info(f"✓ Successfully read from catalog table")
        return df
    except Exception as catalog_error:
        logger.warning(f"Catalog read failed: {str(catalog_error)}")
        logger.info(f"Attempting to read from local file: {local_file_path}")
        try:
            df = spark.read.format("csv")                 .option("header", "true")                 .option("inferSchema", "true")                 .load(local_file_path)
            logger.info(f"✓ Successfully read from local file")
            return df
        except Exception as file_error:
            logger.error(f"Both catalog and file read failed")
            raise Exception(f"Could not read data. Catalog error: {catalog_error}, File error: {file_error}")

# ============================================================================
# NODE IMPLEMENTATIONS
# ============================================================================

# ============================================================================
# NODE node_224: $WORKDB_BILLING.TZ4WN_INCR (Source)
# Container: Batch 1 (Active)
# Description: Source table containing incremental data for external systems.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_224: $WORKDB_BILLING.TZ4WN_INCR")
    
    # Define source table name in Unity Catalog
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TZ4WN_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4WN_INCR.csv"  # Fallback path if catalog fails
    
    # Read data using the helper function
    final_df_node_224 = read_table(table_name, local_path)
    
    # Safe logging of DataFrame information
    log_df_info(final_df_node_224, "Source Node node_224: $WORKDB_BILLING.TZ4WN_INCR")
    
    logger.info("✓ Node node_224 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_224 ($WORKDB_BILLING.TZ4WN_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_224: {str(e)}")
# ============================================================================
# NODE node_225: $NDW_XT.BIL_ExternalSystemType_Id_Xt (Source)
# Container: Batch 1 | File: app_BLBIS_BILL3.BLBIS_BIL_Id_XT1_Analysis (regular)
# Description: Source table used to retrieve the maximum ExternalSystemType_Id 
#              and filter existing ExternalSystemType_Cd values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_225: $NDW_XT.BIL_ExternalSystemType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_externalsystemtype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/ndw_xt_bil_externalsystemtype_id_xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_225 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ExternalSystemType_Id_Xt - Node node_225 ===")
    log_df_info(final_df_node_225, "Source Data Loaded")
    
    logger.info("✓ Node node_225 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_225 ($NDW_XT.BIL_ExternalSystemType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_225: {str(e)}")
# ============================================================================
# NODE node_226: $WORKDB_BILLING.TW5N0_INCR (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing incremental data for processing.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_226: $WORKDB_BILLING.TW5N0_INCR")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "$WORKDB_BILLING.TW5N0_INCR"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5N0_INCR.csv"  # Adjust path if needed
    
    # Read data using the universal read_table helper
    final_df_node_226 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $WORKDB_BILLING.TW5N0_INCR - Node node_226 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_226.columns)}")
    final_df_node_226.printSchema()
    final_df_node_226.show(3, truncate=False)
    
    logger.info("✓ Node node_226 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_226 ($WORKDB_BILLING.TW5N0_INCR): {str(e)}")
    raise Exception(f"Critical error in Node node_226: {str(e)}")
# ============================================================================
# NODE node_227: $NDW_XT.BIL_ExternalSystemType_Id_Xt (Source)
# Container: Batch 1 (Active)
# Description: Source table used to fetch the maximum ExternalSystemType_Id and filter existing SYSTEM_CODE values.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_227: $NDW_XT.BIL_ExternalSystemType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai_powerbi_catalog"
    source_schema = "ndw_xt"
    source_table = "bil_externalsystemtype_id_xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable, for testing purposes)
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_227 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ExternalSystemType_Id_Xt - Node node_227 ===")
    logger.info(f"   └─ Columns: {len(final_df_node_227.columns)}")
    final_df_node_227.printSchema()
    final_df_node_227.show(3, truncate=False)
    
    logger.info("✓ Node node_227 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_227 ($NDW_XT.BIL_ExternalSystemType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_227: {str(e)}")
# ============================================================================
# NODE node_228: $NDW_XT.BIL_ExternalSystemType_Id_Xt (Source)
# Container: genai-powerbi_catalog (Active)
# Description: Source table containing external system type data to be inserted into the target table.
# Previous Node: None
# Next Node: None
# ============================================================================

try:
    logger.info("Processing Node node_228: $NDW_XT.BIL_ExternalSystemType_Id_Xt")
    
    # Define source table name in Unity Catalog
    source_catalog = "genai-powerbi_catalog"
    source_schema = "default"
    source_table = "BIL_ExternalSystemType_Id_Xt"
    table_name = f"{source_catalog}.{source_schema}.{source_table}"
    
    # Define fallback local file path (if applicable)
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"  # Adjust path if necessary
    
    # Read data from Unity Catalog or fallback to local file
    final_df_node_228 = read_table(table_name, local_file_path)
    
    # Safe logging of DataFrame information
    logger.info(f"=== Source: $NDW_XT.BIL_ExternalSystemType_Id_Xt - Node node_228 ===")
    log_df_info(final_df_node_228, "Source DataFrame Information")
    
    logger.info("✓ Node node_228 completed successfully")
    
except Exception as e:
    logger.error(f"✗ Error in Node node_228 ($NDW_XT.BIL_ExternalSystemType_Id_Xt): {str(e)}")
    raise Exception(f"Critical error in Node node_228: {str(e)}")
# ============================================================================



# ======================================================================
# AUTO-GENERATED UNIT TESTS
# Generated by PySparkUnitTestGenAgent (Batch Processing)
# Processed 334 chunks in 34 batches
# ======================================================================

# To run: pytest <this_file>.py -v


import pytest
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql import functions as F
from pyspark.sql.functions import col, lit, when, count, sum as _sum
from chispa.dataframe_comparer import assert_df_equality
from unittest.mock import Mock, patch, MagicMock, call, PropertyMock
from decimal import Decimal
from datetime import datetime, date
import json


@pytest.fixture(scope="session")
def spark_session():
    """Create a SparkSession for testing"""
    spark = SparkSession.builder \
        .master("local[1]") \
        .appName("pytest-pyspark-tests") \
        .config("spark.driver.host", "localhost") \
        .config("spark.sql.shuffle.partitions", "1") \
        .config("spark.sql.adaptive.enabled", "false") \
        .config("spark.ui.enabled", "false") \
        .getOrCreate()
    yield spark
    spark.stop()

@pytest.fixture
def sample_dataframe(spark_session):
    """Fixture providing a sample DataFrame for testing"""
    data = [
        (1, "Alice", 100.0, "A", datetime(2024, 1, 1)),
        (2, "Bob", 200.5, "B", datetime(2024, 1, 2)),
        (3, "Charlie", 300.75, "A", datetime(2024, 1, 3)),
        (4, "David", 400.25, "C", datetime(2024, 1, 4)),
        (5, "Eve", 500.0, "B", datetime(2024, 1, 5))
    ]
    schema = StructType([
        StructField("id", LongType(), True),
        StructField("name", StringType(), True),
        StructField("value", DoubleType(), True),
        StructField("category", StringType(), True),
        StructField("date", TimestampType(), True)
    ])
    return spark_session.createDataFrame(data, schema)

@pytest.fixture
def empty_dataframe(spark_session):
    """Fixture providing an empty DataFrame with schema"""
    schema = StructType([
        StructField("id", LongType(), True),
        StructField("name", StringType(), True)
    ])
    return spark_session.createDataFrame([], schema)


# ======================================================================
# Tests for: Chunk 1 (Lines 31-70)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame"""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame"""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Test")
        mock_logger.assert_any_call("=== Empty DataFrame Test ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception"""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger, patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        log_df_info(df, "Exception Test")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds"""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([("Alice", 1)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds"""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([("Bob", 2)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

# ======================================================================
# Tests for: Chunk 3 (Lines 91-130)
# ======================================================================

def test_read_table_catalog_failure_local_failure(spark_session):
    """Test read_table when both Unity Catalog and local file read fail"""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table("catalog.schema.table", "/path/to/local.csv")
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

def test_log_df_info_valid_dataframe_node_0(spark_session):
    """Test log_df_info for Node 0 with a valid DataFrame"""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Source Node node_0")
        mock_logger.assert_any_call("=== Source Node node_0 ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

# ======================================================================
# Tests for: Chunk 5 (Lines 151-190)
# ======================================================================

def test_read_table_node_1_catalog_success(spark_session):
    """Test read_table for Node 1 when Unity Catalog read succeeds"""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([("Alice", 1)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

def test_read_table_node_1_catalog_failure_local_success(spark_session):
    """Test read_table for Node 1 when Unity Catalog read fails but local file read succeeds"""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([("Bob", 2)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

# ======================================================================
# Tests for: Chunk 6 (Lines 181-220)
# ======================================================================

def test_log_df_info_valid_dataframe_node_2(spark_session):
    """Test log_df_info for Node 2 with a valid DataFrame"""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Source Node node_2")
        mock_logger.assert_any_call("=== Source Node node_2 ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

# ======================================================================
# Tests for: Chunk 7 (Lines 211-250)
# ======================================================================

def test_read_table_node_3_catalog_failure_local_failure(spark_session):
    """Test read_table for Node 3 when both Unity Catalog and local file read fail"""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table("catalog.schema.table", "/path/to/local.csv")
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 8 (Lines 241-280)
# ======================================================================

def test_log_df_info_valid_dataframe_node_4(spark_session):
    """Test log_df_info for Node 4 with a valid DataFrame"""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Source Node node_4")
        mock_logger.assert_any_call("=== Source Node node_4 ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

# ======================================================================
# Tests for: Chunk 9 (Lines 271-310)
# ======================================================================

def test_read_table_node_5_catalog_success(spark_session):
    """Test read_table for Node 5 when Unity Catalog read succeeds"""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([("Alice", 1)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

def test_read_table_node_5_catalog_failure_local_success(spark_session):
    """Test read_table for Node 5 when Unity Catalog read fails but local file read succeeds"""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([("Bob", 2)], ["name", "id"])
        df = read_table("catalog.schema.table", "/path/to/local.csv")
        assert df.count() == 1
        assert df.columns == ["name", "id"]

# ======================================================================
# Tests for: Chunk 11 (Lines 331-370)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame."""
    try:
        log_df_info(None, "Test Step")
    except Exception as e:
        assert str(e) == "'NoneType' object has no attribute 'columns'"

def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 12 (Lines 361-400)
# ======================================================================

def test_read_table_catalog_failure_local_failure(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file reads fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error"))
    try:
        read_table("catalog.schema.table", "/path/to/local.csv")
    except Exception as e:
        assert "Catalog error" in str(e)
        assert "File error" in str(e)

def test_read_table_local_file_success(spark_session, mocker):
    """Test read_table with successful local file read."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_empty_local_file(spark_session, mocker):
    """Test read_table with an empty local file."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 0

# ======================================================================
# Tests for: Chunk 13 (Lines 391-430)
# ======================================================================

def test_node_6_processing_success(spark_session, mocker):
    """Test successful processing of Node node_6."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_6 = read_table("genai_powerbi_catalog.workdb_billing_ta205_incr", "/mnt/data/workdb_billing_ta205_incr.csv")
    assert final_df_node_6.count() == 1
    assert final_df_node_6.columns == ["id", "name"]

def test_node_6_processing_catalog_failure(spark_session, mocker):
    """Test Node node_6 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_6 = read_table("genai_powerbi_catalog.workdb_billing_ta205_incr", "/mnt/data/workdb_billing_ta205_incr.csv")
    assert final_df_node_6.count() == 1
    assert final_df_node_6.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 14 (Lines 421-460)
# ======================================================================

def test_node_7_processing_success(spark_session, mocker):
    """Test successful processing of Node node_7."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_7 = read_table("genai-powerbi_catalog.NDW_XT.BIL_DiaryTransactionType_Id_XT", "/mnt/data/$NDW_XT_BIL_DiaryTransactionType_Id_XT.csv")
    assert final_df_node_7.count() == 1
    assert final_df_node_7.columns == ["id", "name"]

def test_node_7_processing_catalog_failure(spark_session, mocker):
    """Test Node node_7 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_7 = read_table("genai-powerbi_catalog.NDW_XT.BIL_DiaryTransactionType_Id_XT", "/mnt/data/$NDW_XT_BIL_DiaryTransactionType_Id_XT.csv")
    assert final_df_node_7.count() == 1
    assert final_df_node_7.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 15 (Lines 451-490)
# ======================================================================

def test_node_8_processing_success(spark_session, mocker):
    """Test successful processing of Node node_8."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_8 = read_table("genai_powerbi_catalog.ndw_xt.bil_diarytransactiontype_id_xt", "/mnt/data/ndw_xt_bil_diarytransactiontype_id_xt.csv")
    assert final_df_node_8.count() == 1
    assert final_df_node_8.columns == ["id", "name"]

def test_node_8_processing_catalog_failure(spark_session, mocker):
    """Test Node node_8 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_8 = read_table("genai_powerbi_catalog.ndw_xt.bil_diarytransactiontype_id_xt", "/mnt/data/ndw_xt_bil_diarytransactiontype_id_xt.csv")
    assert final_df_node_8.count() == 1
    assert final_df_node_8.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 16 (Lines 481-520)
# ======================================================================

def test_node_9_processing_success(spark_session, mocker):
    """Test successful processing of Node node_9."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_9 = read_table("genai_powerbi_catalog.ndw_type.DiaryTransactionType", None)
    assert final_df_node_9.count() == 1
    assert final_df_node_9.columns == ["id", "name"]

def test_node_9_processing_catalog_failure(spark_session, mocker):
    """Test Node node_9 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    try:
        read_table("genai_powerbi_catalog.ndw_type.DiaryTransactionType", None)
    except Exception as e:
        assert "Catalog error" in str(e)

# ======================================================================
# Tests for: Chunk 17 (Lines 511-550)
# ======================================================================

def test_node_10_processing_success(spark_session, mocker):
    """Test successful processing of Node node_10."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_10 = read_table("genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR", "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv")
    assert final_df_node_10.count() == 1
    assert final_df_node_10.columns == ["id", "name"]

def test_node_10_processing_catalog_failure(spark_session, mocker):
    """Test Node node_10 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_10 = read_table("genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR", "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv")
    assert final_df_node_10.count() == 1
    assert final_df_node_10.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 18 (Lines 541-580)
# ======================================================================

def test_node_11_processing_success(spark_session, mocker):
    """Test successful processing of Node node_11."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_11 = read_table("genai_powerbi_catalog.ndw_xt.bil_finantransstatustype_id_xt", "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv")
    assert final_df_node_11.count() == 1
    assert final_df_node_11.columns == ["id", "name"]

def test_node_11_processing_catalog_failure(spark_session, mocker):
    """Test Node node_11 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_11 = read_table("genai_powerbi_catalog.ndw_xt.bil_finantransstatustype_id_xt", "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv")
    assert final_df_node_11.count() == 1
    assert final_df_node_11.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 19 (Lines 571-610)
# ======================================================================

def test_node_12_processing_success(spark_session, mocker):
    """Test successful processing of Node node_12."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_12 = read_table("genai-powerbi_catalog.$NDW_XT.BIL_FinanTransStatusType_Id_XT", "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv")
    assert final_df_node_12.count() == 1
    assert final_df_node_12.columns == ["id", "name"]

def test_node_12_processing_catalog_failure(spark_session, mocker):
    """Test Node node_12 processing when Unity Catalog read fails."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    final_df_node_12 = read_table("genai-powerbi_catalog.$NDW_XT.BIL_FinanTransStatusType_Id_XT", "/mnt/data/$NDW_XT_BIL_FinanTransStatusType_Id_XT.csv")
    assert final_df_node_12.count() == 1
    assert final_df_node_12.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 21 (Lines 631-670)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test reading from Unity Catalog successfully."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        result = read_table(table_name, local_file_path)
        assert result.count() == 0
        mock_table.assert_called_once_with(table_name)

def test_read_table_catalog_failure_local_success(spark_session):
    """Test fallback to local file when Unity Catalog read fails."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table, \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        result = read_table(table_name, local_file_path)
        assert result.count() == 0
        mock_table.assert_called_once_with(table_name)
        mock_load.assert_called_once_with(local_file_path)

def test_read_table_both_failures(spark_session):
    """Test exception raised when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table, \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")) as mock_load:
        with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
            read_table(table_name, local_file_path)
        mock_table.assert_called_once_with(table_name)
        mock_load.assert_called_once_with(local_file_path)

def test_read_table_local_file_no_path(spark_session):
    """Test exception raised when local file path is None."""
    table_name = "test_catalog.table"
    local_file_path = None
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table:
        with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: 'NoneType' object has no attribute 'load'"):
            read_table(table_name, local_file_path)
        mock_table.assert_called_once_with(table_name)

# ======================================================================
# Tests for: Chunk 24 (Lines 721-760)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test reading from Unity Catalog successfully."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        result = read_table(table_name, local_file_path)
        assert result.count() == 0
        mock_table.assert_called_once_with(table_name)

def test_read_table_catalog_failure_local_success(spark_session):
    """Test fallback to local file when Unity Catalog read fails."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table, \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        result = read_table(table_name, local_file_path)
        assert result.count() == 0
        mock_table.assert_called_once_with(table_name)
        mock_load.assert_called_once_with(local_file_path)

def test_read_table_both_failures(spark_session):
    """Test exception raised when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table, \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")) as mock_load:
        with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
            read_table(table_name, local_file_path)
        mock_table.assert_called_once_with(table_name)
        mock_load.assert_called_once_with(local_file_path)

def test_read_table_local_file_no_path(spark_session):
    """Test exception raised when local file path is None."""
    table_name = "test_catalog.table"
    local_file_path = None
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")) as mock_table:
        with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: 'NoneType' object has no attribute 'load'"):
            read_table(table_name, local_file_path)
        mock_table.assert_called_once_with(table_name)

# ======================================================================
# Tests for: Chunk 37 (Lines 1111-1150)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger:
            log_df_info(df, "Exception Step")
            mock_logger.assert_any_call("=== Exception Step ===")
            mock_logger.assert_any_call("   └─ Columns: 2")
            mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 38 (Lines 1141-1180)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result.count() == 2
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result.count() == 2
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/local.csv")
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 39 (Lines 1171-1210)
# ======================================================================

def test_node_26_processing_success(spark_session):
    """Test Node 26 processing when data is successfully read."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df):
        with patch("logging.Logger.info") as mock_logger:
            final_df_node_26 = read_table("genai-powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv")
            log_df_info(final_df_node_26, "Source Node node_26: $WORKDB_BILLING.TZ4A0_INCR")
            assert final_df_node_26.count() == 2
            assert final_df_node_26.columns == ["id", "name"]
            mock_logger.assert_any_call("✓ Node node_26 completed successfully")

def test_node_26_catalog_failure_local_success(spark_session):
    """Test Node 26 processing when catalog read fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df):
            final_df_node_26 = read_table("genai-powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv")
            assert final_df_node_26.count() == 2
            assert final_df_node_26.columns == ["id", "name"]

def test_node_26_both_failures(spark_session):
    """Test Node 26 processing when both catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("genai-powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv")
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 43 (Lines 1291-1330)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Region1"), (2, "Region2"), (3, "Region3")]
    schema = StructType([
        StructField("RegionType_Id", IntegerType(), True),
        StructField("RegionName", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("RegionType_Id", IntegerType(), True),
        StructField("RegionName", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when show() raises an exception."""
    test_data = [(1, "Region1"), (2, "Region2"), (3, "Region3")]
    schema = StructType([
        StructField("RegionType_Id", IntegerType(), True),
        StructField("RegionName", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger:
            log_df_info(df, "Error Step")
            mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 44 (Lines 1321-1360)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Region1"), (2, "Region2")]
    schema = StructType([
        StructField("RegionType_Id", IntegerType(), True),
        StructField("RegionName", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert_df_equality(result, df)

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    test_data = [(1, "Region1"), (2, "Region2")]
    schema = StructType([
        StructField("RegionType_Id", IntegerType(), True),
        StructField("RegionName", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert_df_equality(result, df)

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 54 (Lines 1621-1660)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")

def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame (None)."""
    with pytest.raises(AttributeError):
        log_df_info(None, "Test Step")

def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 55 (Lines 1651-1690)
# ======================================================================

def test_read_table_catalog_and_local_failure(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error"))
    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
        read_table("catalog.schema.table", "/path/to/local.csv")

def test_read_table_local_file_success(spark_session, mocker):
    """Test read_table with a valid local file path."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_empty_local_file(spark_session, mocker):
    """Test read_table with an empty local file."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([], StructType([])))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 0

# ======================================================================
# Tests for: Chunk 57 (Lines 1711-1750)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)
    
    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")

def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame (None)."""
    with pytest.raises(AttributeError):
        log_df_info(None, "Test Step")

def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 58 (Lines 1741-1780)
# ======================================================================

def test_read_table_catalog_and_local_failure(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error"))
    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
        read_table("catalog.schema.table", "/path/to/local.csv")

def test_read_table_local_file_success(spark_session, mocker):
    """Test read_table with a valid local file path."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_empty_local_file(spark_session, mocker):
    """Test read_table with an empty local file."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([], StructType([])))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 0

# ======================================================================
# Tests for: Chunk 61 (Lines 1831-1870)
# ======================================================================

def test_read_table_valid_input(spark_session, mocker):
    """Test read_table with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementStatusType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table(spark_session, mocker):
    """Test read_table with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementStatusType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path(spark_session, mocker):
    """Test read_table with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementStatusType_Id_XT"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 62 (Lines 1861-1900)
# ======================================================================

def test_read_table_valid_input_node_42(spark_session, mocker):
    """Test read_table for Node 42 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.default.bil_agreementstatustype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_42(spark_session, mocker):
    """Test read_table for Node 42 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.default.bil_agreementstatustype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementStatusType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_42(spark_session, mocker):
    """Test read_table for Node 42 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.default.bil_agreementstatustype_id_xt"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 63 (Lines 1891-1930)
# ======================================================================

def test_read_table_valid_input_node_43(spark_session, mocker):
    """Test read_table for Node 43 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementStatusType"
    local_path = "/mnt/data/ndw_type/AgreementStatusType.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_43(spark_session, mocker):
    """Test read_table for Node 43 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementStatusType"
    local_path = "/mnt/data/ndw_type/AgreementStatusType.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_43(spark_session, mocker):
    """Test read_table for Node 43 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementStatusType"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 64 (Lines 1921-1960)
# ======================================================================

def test_read_table_valid_input_node_44(spark_session, mocker):
    """Test read_table for Node 44 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_44(spark_session, mocker):
    """Test read_table for Node 44 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ4A0_INCR.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_44(spark_session, mocker):
    """Test read_table for Node 44 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 65 (Lines 1951-1990)
# ======================================================================

def test_read_table_valid_input_node_45(spark_session, mocker):
    """Test read_table for Node 45 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_45(spark_session, mocker):
    """Test read_table for Node 45 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_45(spark_session, mocker):
    """Test read_table for Node 45 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_AgreementType_Id_XT"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 66 (Lines 1981-2020)
# ======================================================================

def test_read_table_valid_input_node_46(spark_session, mocker):
    """Test read_table for Node 46 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.default.BIL_AgreementType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_46(spark_session, mocker):
    """Test read_table for Node 46 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.default.BIL_AgreementType_Id_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_AgreementType_Id_XT.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_46(spark_session, mocker):
    """Test read_table for Node 46 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.default.BIL_AgreementType_Id_XT"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 67 (Lines 2011-2050)
# ======================================================================

def test_read_table_valid_input_node_47(spark_session, mocker):
    """Test read_table for Node 47 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementType"
    local_path = "/mnt/data/ndw_type_agreementtype.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_47(spark_session, mocker):
    """Test read_table for Node 47 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementType"
    local_path = "/mnt/data/ndw_type_agreementtype.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_47(spark_session, mocker):
    """Test read_table for Node 47 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.ndw_type.AgreementType"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 68 (Lines 2041-2080)
# ======================================================================

def test_read_table_valid_input_node_48(spark_session, mocker):
    """Test read_table for Node 48 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5JB_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5JB_INCR.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_48(spark_session, mocker):
    """Test read_table for Node 48 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5JB_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5JB_INCR.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_48(spark_session, mocker):
    """Test read_table for Node 48 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai_powerbi_catalog.$WORKDB_BILLING.TW5JB_INCR"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 69 (Lines 2071-2110)
# ======================================================================

def test_read_table_valid_input_node_49(spark_session, mocker):
    """Test read_table for Node 49 with valid Unity Catalog table name and fallback path."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BankAccountType_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_BankAccountType_Id_Xt.csv"
    
    result_df = read_table(table_name, local_path)
    
    mock_read_table.assert_called_once_with(table_name, local_path)
    assert result_df.count() == 0

def test_read_table_empty_table_node_49(spark_session, mocker):
    """Test read_table for Node 49 with an empty DataFrame."""
    mock_read_table = mocker.patch("read_table", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BankAccountType_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_BankAccountType_Id_Xt.csv"
    
    result_df = read_table(table_name, local_path)
    
    assert result_df.count() == 0
    assert len(result_df.columns) == 0

def test_read_table_invalid_path_node_49(spark_session, mocker):
    """Test read_table for Node 49 with an invalid fallback path."""
    mock_read_table = mocker.patch("read_table", side_effect=Exception("Invalid path"))
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_BankAccountType_Id_Xt"
    local_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="Invalid path"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 75 (Lines 2251-2290)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame."""
    try:
        log_df_info(None, "Test Step")
    except Exception as e:
        assert str(e) == "'NoneType' object has no attribute 'columns'"

def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error"))
    with pytest.raises(Exception) as excinfo:
        read_table("catalog.schema.table", "/path/to/local.csv")
    assert "Catalog error" in str(excinfo.value)
    assert "File error" in str(excinfo.value)

# ======================================================================
# Tests for: Chunk 80 (Lines 2401-2440)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Test Step")
        mock_logger_info.assert_any_call("=== Test Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Empty Step")
        mock_logger_info.assert_any_call("=== Empty Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")), \
         patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Error Step")
        mock_logger_info.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 81 (Lines 2431-2470)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 82 (Lines 2461-2500)
# ======================================================================

def test_node_58_processing_success(spark_session):
    """Test Node 58 processing when read_table succeeds."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        final_df_node_58 = read_table(table_name, local_path)
        log_df_info(final_df_node_58, "Source Node node_58: $WORKDB_BILLING.TZ5A0_INCR")
        assert final_df_node_58.count() == 0

def test_node_58_processing_catalog_failure_local_success(spark_session):
    """Test Node 58 processing when Unity Catalog fails but local file read succeeds."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        final_df_node_58 = read_table(table_name, local_path)
        log_df_info(final_df_node_58, "Source Node node_58: $WORKDB_BILLING.TZ5A0_INCR")
        assert final_df_node_58.count() == 0

def test_node_58_processing_both_failures(spark_session):
    """Test Node 58 processing when both Unity Catalog and local file read fail."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 105 (Lines 3151-3190)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")


def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")


def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame."""
    df = None  # Invalid DataFrame

    with pytest.raises(AttributeError):
        log_df_info(df, "Test Step")


def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"
    df = read_table(table_name, local_file_path)

    assert df.count() == 1
    assert df.columns == ["id", "name"]


def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"
    df = read_table(table_name, local_file_path)

    assert df.count() == 1
    assert df.columns == ["id", "name"]


def test_read_table_both_failures(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with pytest.raises(Exception, match="Could not read data"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 106 (Lines 3181-3220)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")


def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        pytest.fail(f"Unexpected exception: {str(e)}")


def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame."""
    df = None  # Invalid DataFrame

    with pytest.raises(AttributeError):
        log_df_info(df, "Test Step")


def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"
    df = read_table(table_name, local_file_path)

    assert df.count() == 1
    assert df.columns == ["id", "name"]


def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"
    df = read_table(table_name, local_file_path)

    assert df.count() == 1
    assert df.columns == ["id", "name"]


def test_read_table_both_failures(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with pytest.raises(Exception, match="Could not read data"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 111 (Lines 3331-3370)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Test Step")
        mock_logger_info.assert_any_call("=== Test Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger_info.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_error(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger_info:
            log_df_info(df, "Show Error Step")
            mock_logger_info.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 112 (Lines 3361-3400)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([], StructType([]))
            df = read_table(table_name, local_file_path)
            assert df.count() == 0

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 119 (Lines 3571-3610)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Test Step")
        mock_logger_info.assert_any_call("=== Test Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger_info.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_error(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger_info:
            log_df_info(df, "Show Error Step")
            mock_logger_info.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([], StructType([]))
            df = read_table(table_name, local_file_path)
            assert df.count() == 0

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 124 (Lines 3721-3760)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty Step")
        mock_logger.assert_any_call("=== Empty Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger:
            log_df_info(df, "Error Step")
            mock_logger.assert_any_call("=== Error Step ===")
            mock_logger.assert_any_call("   └─ Columns: 2")
            mock_logger.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df):
        result = read_table("catalog.schema.table", None)
        assert result.count() == 2
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df):
            result = read_table("catalog.schema.table", "/path/to/file.csv")
            assert result.count() == 2
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/file.csv")
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 130 (Lines 3901-3940)
# ======================================================================

from pyspark.sql import DataFrame

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Test Step")
        mock_logger_info.assert_any_call("=== Test Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger_info:
        log_df_info(df, "Empty Step")
        mock_logger_info.assert_any_call("=== Empty Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger_info:
            log_df_info(df, "Error Step")
            mock_logger_info.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 131 (Lines 3931-3970)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert isinstance(df, DataFrame)
        assert df.count() == 1

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
            df = read_table(table_name, local_file_path)
            assert isinstance(df, DataFrame)
            assert df.count() == 1

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 132 (Lines 3961-4000)
# ======================================================================

def test_node_93_processing_success(spark_session):
    """Test Node 93 processing with successful data read."""
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_dunningcodetype_id_xt.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_93 = read_table(table_name, local_path)
        assert final_df_node_93.count() == 1

def test_node_93_processing_failure(spark_session):
    """Test Node 93 processing when data read fails."""
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_path = "/mnt/data/ndw_xt_bil_dunningcodetype_id_xt.csv"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert str(e) == "Critical error in Node node_93: Read error"

# ======================================================================
# Tests for: Chunk 133 (Lines 3991-4030)
# ======================================================================

def test_node_94_processing_success(spark_session):
    """Test Node 94 processing with successful data read."""
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_DunningCodeType_Id_XT.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_94 = read_table(table_name, local_file_path)
        assert final_df_node_94.count() == 1

def test_node_94_processing_failure(spark_session):
    """Test Node 94 processing when data read fails."""
    table_name = "genai-powerbi_catalog.ndw_xt.bil_dunningcodetype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_DunningCodeType_Id_XT.csv"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Critical error in Node node_94: Read error"

# ======================================================================
# Tests for: Chunk 134 (Lines 4021-4060)
# ======================================================================

def test_node_95_processing_success(spark_session):
    """Test Node 95 processing with successful data read."""
    table_name = "genai-powerbi_catalog.ndw_type.dunningcodetype"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_95 = read_table(table_name, None)
        assert final_df_node_95.count() == 1

def test_node_95_processing_failure(spark_session):
    """Test Node 95 processing when data read fails."""
    table_name = "genai-powerbi_catalog.ndw_type.dunningcodetype"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(table_name, None)
        except Exception as e:
            assert str(e) == "Critical error in Node node_95: Read error"

# ======================================================================
# Tests for: Chunk 137 (Lines 4111-4150)
# ======================================================================

def test_node_96_processing_success(spark_session):
    """Test Node 96 processing with successful data read."""
    table_name = "genai-powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_96 = read_table(table_name, local_path)
        assert final_df_node_96.count() == 1

def test_node_96_processing_failure(spark_session):
    """Test Node 96 processing when data read fails."""
    table_name = "genai-powerbi_catalog.workdb_billing_tz5a0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TZ5A0_INCR.csv"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert str(e) == "Critical error in Node node_96: Read error"

# ======================================================================
# Tests for: Chunk 138 (Lines 4141-4180)
# ======================================================================

def test_node_97_processing_success(spark_session):
    """Test Node 97 processing with successful data read."""
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_EFTStatementType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_97 = read_table(source_table_name, local_file_path)
        assert final_df_node_97.count() == 1

def test_node_97_processing_failure(spark_session):
    """Test Node 97 processing when data read fails."""
    source_table_name = "genai-powerbi_catalog.$NDW_XT.BIL_EFTStatementType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(source_table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Critical error in Node node_97: Read error"

# ======================================================================
# Tests for: Chunk 139 (Lines 4171-4210)
# ======================================================================

def test_node_98_processing_success(spark_session):
    """Test Node 98 processing with successful data read."""
    table_name = "genai-powerbi_catalog.default.$NDW_XT.BIL_EFTStatementType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_98 = read_table(table_name, local_file_path)
        assert final_df_node_98.count() == 1

def test_node_98_processing_failure(spark_session):
    """Test Node 98 processing when data read fails."""
    table_name = "genai-powerbi_catalog.default.$NDW_XT.BIL_EFTStatementType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTStatementType_Id_Xt.csv"

    with patch("read_table", side_effect=Exception("Read error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Critical error in Node node_98: Read error"

# ======================================================================
# Tests for: Chunk 141 (Lines 4231-4270)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty Step")
        mock_logger.assert_any_call("=== Empty Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_error(spark_session):
    """Test log_df_info when show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")), \
         patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Error Step")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 142 (Lines 4261-4300)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table successfully reads from Unity Catalog."""
    table_name = "catalog.schema.table"
    local_file_path = None

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table falls back to local file when catalog read fails."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(2, "Bob")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table raises an exception when both catalog and local file reads fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 143 (Lines 4291-4330)
# ======================================================================

def test_node_100_processing_success(spark_session):
    """Test Node 100 processing with valid data."""
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TU5P0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TU5P0_INCR.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_100 = read_table(table_name, local_path)
        assert final_df_node_100.count() == 1
        assert final_df_node_100.columns == ["id", "name"]

def test_node_100_processing_catalog_failure(spark_session):
    """Test Node 100 processing when catalog read fails."""
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TU5P0_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TU5P0_INCR.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 144 (Lines 4321-4360)
# ======================================================================

def test_node_101_processing_success(spark_session):
    """Test Node 101 processing with valid data."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_efttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT.BIL_EFTType_Id_XT.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_101 = read_table(table_name, local_file_path)
        assert final_df_node_101.count() == 1
        assert final_df_node_101.columns == ["id", "name"]

def test_node_101_processing_catalog_failure(spark_session):
    """Test Node 101 processing when catalog read fails."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_efttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT.BIL_EFTType_Id_XT.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 145 (Lines 4351-4390)
# ======================================================================

def test_node_102_processing_success(spark_session):
    """Test Node 102 processing with valid data."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_efttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTType_Id_XT.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_102 = read_table(table_name, local_file_path)
        assert final_df_node_102.count() == 1
        assert final_df_node_102.columns == ["id", "name"]

def test_node_102_processing_catalog_failure(spark_session):
    """Test Node 102 processing when catalog read fails."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_efttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_EFTType_Id_XT.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 146 (Lines 4381-4420)
# ======================================================================

def test_node_103_processing_success(spark_session):
    """Test Node 103 processing with valid data."""
    table_name = "genai_powerbi_catalog.ndw_type.ElectronicFundTransType"
    local_file_path = "/mnt/data/ndw_type/ElectronicFundTransType.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_103 = read_table(table_name, local_file_path)
        assert final_df_node_103.count() == 1
        assert final_df_node_103.columns == ["id", "name"]

def test_node_103_processing_catalog_failure(spark_session):
    """Test Node 103 processing when catalog read fails."""
    table_name = "genai_powerbi_catalog.ndw_type.ElectronicFundTransType"
    local_file_path = "/mnt/data/ndw_type/ElectronicFundTransType.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 147 (Lines 4411-4450)
# ======================================================================

def test_node_104_processing_success(spark_session):
    """Test Node 104 processing with valid data."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw6g0_incr"
    local_file_path = None

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_104 = read_table(table_name, local_file_path)
        assert final_df_node_104.count() == 1
        assert final_df_node_104.columns == ["id", "name"]

def test_node_104_processing_catalog_failure(spark_session):
    """Test Node 104 processing when catalog read fails."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw6g0_incr"
    local_file_path = None

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 148 (Lines 4441-4480)
# ======================================================================

def test_node_105_processing_success(spark_session):
    """Test Node 105 processing with valid data."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgeraccounttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_GenLedgerAccountType_Id_XT.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_105 = read_table(table_name, local_file_path)
        assert final_df_node_105.count() == 1
        assert final_df_node_105.columns == ["id", "name"]

def test_node_105_processing_catalog_failure(spark_session):
    """Test Node 105 processing when catalog read fails."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgeraccounttype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_GenLedgerAccountType_Id_XT.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 149 (Lines 4471-4510)
# ======================================================================

def test_node_106_processing_success(spark_session):
    """Test Node 106 processing with valid data."""
    table_name = "genai-powerbi_catalog.default.$NDW_XT.Bil_GenLedgerAccountType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_Bil_GenLedgerAccountType_Id_XT.csv"

    with patch("read_table") as mock_read_table:
        mock_read_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_106 = read_table(table_name, local_file_path)
        assert final_df_node_106.count() == 1
        assert final_df_node_106.columns == ["id", "name"]

def test_node_106_processing_catalog_failure(spark_session):
    """Test Node 106 processing when catalog read fails."""
    table_name = "genai-powerbi_catalog.default.$NDW_XT.Bil_GenLedgerAccountType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_Bil_GenLedgerAccountType_Id_XT.csv"

    with patch("read_table", side_effect=Exception("Catalog error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Catalog error"

# ======================================================================
# Tests for: Chunk 152 (Lines 4561-4600)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Test")
        mock_logger.assert_any_call("=== Empty DataFrame Test ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_error(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")), \
         patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Show Error Test")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 153 (Lines 4591-4630)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 155 (Lines 4651-4690)
# ======================================================================

def test_read_table_node_109_catalog_success(spark_session):
    """Test read_table for Node 109 when Unity Catalog read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_109_catalog_failure_local_success(spark_session):
    """Test read_table for Node 109 when Unity Catalog read fails but local file read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_109_both_failures(spark_session):
    """Test read_table for Node 109 when both Unity Catalog and local file read fail."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT_BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 158 (Lines 4741-4780)
# ======================================================================

def test_read_table_node_110_catalog_success(spark_session):
    """Test read_table for Node 110 when Unity Catalog read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_110_catalog_failure_local_success(spark_session):
    """Test read_table for Node 110 when Unity Catalog read fails but local file read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_110_both_failures(spark_session):
    """Test read_table for Node 110 when both Unity Catalog and local file read fail."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_genledgereventtype_id_xt"
    local_path = "/mnt/data/$NDW_XT.BIL_GenLedgerEventType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 159 (Lines 4771-4810)
# ======================================================================

def test_read_table_node_111_catalog_success(spark_session):
    """Test read_table for Node 111 when Unity Catalog read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_type.GeneralLedgerEventType"
    local_file_path = "/mnt/data/ndw_type/GeneralLedgerEventType.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_111_catalog_failure_local_success(spark_session):
    """Test read_table for Node 111 when Unity Catalog read fails but local file read succeeds."""
    table_name = "genai_powerbi_catalog.ndw_type.GeneralLedgerEventType"
    local_file_path = "/mnt/data/ndw_type/GeneralLedgerEventType.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Event1")], ["id", "event"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert df.columns == ["id", "event"]

def test_read_table_node_111_both_failures(spark_session):
    """Test read_table for Node 111 when both Unity Catalog and local file read fail."""
    table_name = "genai_powerbi_catalog.ndw_type.GeneralLedgerEventType"
    local_file_path = "/mnt/data/ndw_type/GeneralLedgerEventType.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 161 (Lines 4831-4870)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_exception(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger, patch.object(df, "show", side_effect=Exception("Show error")):
        log_df_info(df, "Show Error Step")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"

    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table, patch("logging.Logger.info") as mock_logger:
        result = read_table(table_name, local_file_path)
        mock_table.assert_called_once_with(table_name)
        mock_logger.assert_any_call("✓ Successfully read from catalog table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.table"
    local_file_path = "/path/to/local.csv"

    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load, \
         patch("logging.Logger.info") as mock_logger:
        result = read_table(table_name, local_file_path)
        mock_load.assert_called_once_with(local_file_path)
        mock_logger.assert_any_call("✓ Successfully read from local file")
        assert result == mock_df

# ======================================================================
# Tests for: Chunk 170 (Lines 5101-5140)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 25), ("Bob", 30)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logger.info") as mock_logger_info:
        log_df_info(df, "Test Step")
        mock_logger_info.assert_any_call("=== Test Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logger.info") as mock_logger_info:
        log_df_info(df, "Empty Step")
        mock_logger_info.assert_any_call("=== Empty Step ===")
        mock_logger_info.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_exception(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 25), ("Bob", 30)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logger.info") as mock_logger_info, patch.object(df, "show", side_effect=Exception("Show error")):
        log_df_info(df, "Error Step")
        mock_logger_info.assert_any_call("   └─ Could not show data: Show error")

@patch("spark.table")
def test_read_table_catalog_success(mock_spark_table, spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    mock_spark_table.return_value = mock_df

    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result == mock_df
    mock_spark_table.assert_called_once_with("catalog.schema.table")

@patch("spark.table", side_effect=Exception("Catalog error"))
@patch("spark.read.format().load")
def test_read_table_local_success(mock_spark_read_load, mock_spark_table, spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    mock_df = MagicMock()
    mock_spark_read_load.return_value = mock_df

    result = read_table("catalog.schema.table", "/path/to/local.csv")
    assert result == mock_df
    mock_spark_table.assert_called_once_with("catalog.schema.table")
    mock_spark_read_load.assert_called_once_with("/path/to/local.csv")

@patch("spark.table", side_effect=Exception("Catalog error"))
@patch("spark.read.format().load", side_effect=Exception("File error"))
def test_read_table_both_fail(mock_spark_read_load, mock_spark_table, spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
        read_table("catalog.schema.table", "/path/to/local.csv")
    mock_spark_table.assert_called_once_with("catalog.schema.table")
    mock_spark_read_load.assert_called_once_with("/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 181 (Lines 5431-5470)
# ======================================================================

def test_read_table_node_127_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 127 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Plan A"), (2, "Plan B")],
        ["PlanStatusType_Id", "PlanStatus"]
    ))
    
    table_name = "genai-powerbi_catalog.NDW_XT.BIL_PlanStatusType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "PlanStatusType_Id" in result_df.columns
    assert "PlanStatus" in result_df.columns

def test_read_table_node_127_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 127 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["PlanStatusType_Id", "PlanStatus"]))
    
    table_name = "genai-powerbi_catalog.NDW_XT.BIL_PlanStatusType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_127_invalid_path(spark_session, mocker):
    """Test reading data for Node 127 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.NDW_XT.BIL_PlanStatusType_Id_XT"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 182 (Lines 5461-5500)
# ======================================================================

def test_read_table_node_128_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 128 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Plan A"), (2, "Plan B")],
        ["PlanStatusType_Id", "PlanStatus"]
    ))
    
    table_name = "genai-powerbi_catalog.default.BIL_PlanStatusType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "PlanStatusType_Id" in result_df.columns
    assert "PlanStatus" in result_df.columns

def test_read_table_node_128_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 128 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["PlanStatusType_Id", "PlanStatus"]))
    
    table_name = "genai-powerbi_catalog.default.BIL_PlanStatusType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_PlanStatusType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_128_invalid_path(spark_session, mocker):
    """Test reading data for Node 128 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.default.BIL_PlanStatusType_Id_XT"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 183 (Lines 5491-5530)
# ======================================================================

def test_read_table_node_129_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 129 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Type A"), (2, "Type B")],
        ["ProducerType_Id", "ProducerType"]
    ))
    
    table_name = "genai-powerbi_catalog.ndw_type.PlanStatusType"
    local_file_path = "/mnt/data/ndw_type/PlanStatusType.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "ProducerType_Id" in result_df.columns
    assert "ProducerType" in result_df.columns

def test_read_table_node_129_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 129 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ProducerType_Id", "ProducerType"]))
    
    table_name = "genai-powerbi_catalog.ndw_type.PlanStatusType"
    local_file_path = "/mnt/data/ndw_type/PlanStatusType.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_129_invalid_path(spark_session, mocker):
    """Test reading data for Node 129 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.ndw_type.PlanStatusType"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 184 (Lines 5521-5560)
# ======================================================================

def test_read_table_node_130_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 130 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Category A"), (2, "Category B")],
        ["Category_Id", "Category"]
    ))
    
    table_name = "genai_powerbi_catalog.workdb_billing.tz5bo_incr"
    local_file_path = "/mnt/data/workdb_billing/tz5bo_incr.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "Category_Id" in result_df.columns
    assert "Category" in result_df.columns

def test_read_table_node_130_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 130 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["Category_Id", "Category"]))
    
    table_name = "genai_powerbi_catalog.workdb_billing.tz5bo_incr"
    local_file_path = "/mnt/data/workdb_billing/tz5bo_incr.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_130_invalid_path(spark_session, mocker):
    """Test reading data for Node 130 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai_powerbi_catalog.workdb_billing.tz5bo_incr"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 185 (Lines 5551-5590)
# ======================================================================

def test_read_table_node_131_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 131 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Producer A"), (2, "Producer B")],
        ["ProducerType_Id", "ProducerType"]
    ))
    
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "ProducerType_Id" in result_df.columns
    assert "ProducerType" in result_df.columns

def test_read_table_node_131_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 131 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ProducerType_Id", "ProducerType"]))
    
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_131_invalid_path(spark_session, mocker):
    """Test reading data for Node 131 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 186 (Lines 5581-5620)
# ======================================================================

def test_read_table_node_132_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 132 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Producer A"), (2, "Producer B")],
        ["ProducerType_Id", "ProducerType"]
    ))
    
    table_name = "genai-powerbi_catalog.default.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "ProducerType_Id" in result_df.columns
    assert "ProducerType" in result_df.columns

def test_read_table_node_132_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 132 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ProducerType_Id", "ProducerType"]))
    
    table_name = "genai-powerbi_catalog.default.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ProducerType_Id_XT.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_132_invalid_path(spark_session, mocker):
    """Test reading data for Node 132 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.default.$NDW_XT.BIL_ProducerType_Id_XT"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 187 (Lines 5611-5650)
# ======================================================================

def test_read_table_node_133_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 133 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Producer A"), (2, "Producer B")],
        ["ProducerType_Id", "ProducerType"]
    ))
    
    table_name = "genai-powerbi_catalog.ndw_type.ProducerType"
    result_df = read_table(table_name, local_file_path=None)
    
    assert result_df.count() == 2
    assert "ProducerType_Id" in result_df.columns
    assert "ProducerType" in result_df.columns

def test_read_table_node_133_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 133 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ProducerType_Id", "ProducerType"]))
    
    table_name = "genai-powerbi_catalog.ndw_type.ProducerType"
    result_df = read_table(table_name, local_file_path=None)
    
    assert result_df.count() == 0

def test_read_table_node_133_no_local_path(spark_session, mocker):
    """Test reading data for Node 133 without a local file path fallback."""
    mocker.patch("read_table", side_effect=Exception("No local path provided"))
    
    table_name = "genai-powerbi_catalog.ndw_type.ProducerType"
    
    with pytest.raises(Exception, match="No local path provided"):
        read_table(table_name, local_file_path=None)

# ======================================================================
# Tests for: Chunk 188 (Lines 5641-5680)
# ======================================================================

def test_read_table_node_134_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 134 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Reason A"), (2, "Reason B")],
        ["ReasonCode_Id", "ReasonCode"]
    ))
    
    table_name = "genai-powerbi_catalog.workdb_billing.tw5t2_incr"
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5T2_INCR.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "ReasonCode_Id" in result_df.columns
    assert "ReasonCode" in result_df.columns

def test_read_table_node_134_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 134 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ReasonCode_Id", "ReasonCode"]))
    
    table_name = "genai-powerbi_catalog.workdb_billing.tw5t2_incr"
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5T2_INCR.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_134_invalid_path(spark_session, mocker):
    """Test reading data for Node 134 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.workdb_billing.tw5t2_incr"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 189 (Lines 5671-5710)
# ======================================================================

def test_read_table_node_135_valid_input(spark_session, mocker):
    """Test reading data from Unity Catalog for Node 135 with valid input."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame(
        [(1, "Return A"), (2, "Return B")],
        ["ReasonReturnType_Id", "ReasonReturnType"]
    ))
    
    table_name = "genai-powerbi_catalog.ndw_xt.bil_reasonreturntype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT.BIL_ReasonReturnType_Id_Xt.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 2
    assert "ReasonReturnType_Id" in result_df.columns
    assert "ReasonReturnType" in result_df.columns

def test_read_table_node_135_empty_dataframe(spark_session, mocker):
    """Test reading data for Node 135 when the source table is empty."""
    mocker.patch("read_table", return_value=spark_session.createDataFrame([], ["ReasonReturnType_Id", "ReasonReturnType"]))
    
    table_name = "genai-powerbi_catalog.ndw_xt.bil_reasonreturntype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT.BIL_ReasonReturnType_Id_Xt.csv"
    result_df = read_table(table_name, local_file_path)
    
    assert result_df.count() == 0

def test_read_table_node_135_invalid_path(spark_session, mocker):
    """Test reading data for Node 135 with an invalid local file path."""
    mocker.patch("read_table", side_effect=Exception("File not found"))
    
    table_name = "genai-powerbi_catalog.ndw_xt.bil_reasonreturntype_id_xt"
    local_file_path = "/invalid/path.csv"
    
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 197 (Lines 5911-5950)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger:
            log_df_info(df, "Error Step")
            mock_logger.assert_any_call("=== Error Step ===")
            mock_logger.assert_any_call("   └─ Columns: 2")
            mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 198 (Lines 5941-5980)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result.count() == 2
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result.count() == 2
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/local.csv")
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 201 (Lines 6031-6070)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice",), ("Bob",)]
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 1")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 1")

def test_log_df_info_show_exception(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger, patch.object(df, "show", side_effect=Exception("Show error")):
        log_df_info(df, "Show Error Step")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/mnt/data/test.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/mnt/data/test.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

# ======================================================================
# Tests for: Chunk 207 (Lines 6211-6250)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice",), ("Bob",)]
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 1")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Step")
        mock_logger.assert_any_call("=== Empty DataFrame Step ===")
        mock_logger.assert_any_call("   └─ Columns: 1")

def test_log_df_info_show_exception(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    schema = StructType([StructField("name", StringType(), True)])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger, patch.object(df, "show", side_effect=Exception("Show error")):
        log_df_info(df, "Show Error Step")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/mnt/data/test.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/mnt/data/test.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df.count() == 0

# ======================================================================
# Tests for: Chunk 210 (Lines 6301-6340)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Empty Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked Exception")):
        log_df_info(df, "Exception Step")
        # No exception should be raised, and logger should log the error

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df is not None
        mock_table.assert_called_once_with(table_name)

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([], StructType([]))
            df = read_table(table_name, local_file_path)
            assert df is not None
            mock_load.assert_called_once_with(local_file_path)

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert "Catalog Error" in str(e)
                assert "File Error" in str(e)

# ======================================================================
# Tests for: Chunk 219 (Lines 6571-6610)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Empty Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked Exception")):
        log_df_info(df, "Exception Step")
        # No exception should be raised, and logger should log the error

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([], StructType([]))
        df = read_table(table_name, local_file_path)
        assert df is not None
        mock_table.assert_called_once_with(table_name)

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([], StructType([]))
            df = read_table(table_name, local_file_path)
            assert df is not None
            mock_load.assert_called_once_with(local_file_path)

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert "Catalog Error" in str(e)
                assert "File Error" in str(e)

# ======================================================================
# Tests for: Chunk 221 (Lines 6631-6670)
# ======================================================================

def test_read_table_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_PolicyPhysicalSystem_Id_Xt.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Policy1"), (2, "Policy2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("policy_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "policy_name" in result_df.columns


def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_PolicyPhysicalSystem_Id_Xt.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Policy1"), (2, "Policy2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("policy_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "policy_name" in result_df.columns


def test_read_table_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_PolicyPhysicalSystem_Id_Xt"
    local_path = "/mnt/data/$NDW_XT_BIL_PolicyPhysicalSystem_Id_Xt.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 222 (Lines 6661-6700)
# ======================================================================

def test_read_table_node_157_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 157."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyPhysicalSystemType"
    local_path = "/mnt/data/ndw_type/PolicyPhysicalSystemType.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Type1"), (2, "Type2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "type_name" in result_df.columns


def test_read_table_node_157_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 157."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyPhysicalSystemType"
    local_path = "/mnt/data/ndw_type/PolicyPhysicalSystemType.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Type1"), (2, "Type2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "type_name" in result_df.columns


def test_read_table_node_157_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 157."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyPhysicalSystemType"
    local_path = "/mnt/data/ndw_type/PolicyPhysicalSystemType.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 223 (Lines 6691-6730)
# ======================================================================

def test_read_table_node_158_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 158."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Increment1"), (2, "Increment2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("increment_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "increment_name" in result_df.columns


def test_read_table_node_158_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 158."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Increment1"), (2, "Increment2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("increment_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "increment_name" in result_df.columns


def test_read_table_node_158_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 158."""
    table_name = "genai_powerbi_catalog.workdb_billing_tz4a0_incr"
    local_path = "/mnt/data/workdb_billing_tz4a0_incr.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 224 (Lines 6721-6760)
# ======================================================================

def test_read_table_node_159_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 159."""
    table_name = "genai-powerbi_catalog.default.BIL_PolicyBrand_Id_Xt"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Brand1"), (2, "Brand2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("brand_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = spark_session.table(table_name)

    assert result_df.count() == 2
    assert "brand_name" in result_df.columns


def test_read_table_node_159_catalog_failure(spark_session, mocker):
    """Test exception raised when Unity Catalog read fails for Node 159."""
    table_name = "genai-powerbi_catalog.default.BIL_PolicyBrand_Id_Xt"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    with pytest.raises(Exception, match="Catalog read failed"):
        spark_session.table(table_name)

# ======================================================================
# Tests for: Chunk 225 (Lines 6751-6790)
# ======================================================================

def test_read_table_node_160_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 160."""
    table_name = "genai-powerbi_catalog.default.BIL_PolicyBrand_Id_Xt"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Brand1"), (2, "Brand2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("brand_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_file_path=None)

    assert result_df.count() == 2
    assert "brand_name" in result_df.columns


def test_read_table_node_160_catalog_failure(spark_session, mocker):
    """Test exception raised when Unity Catalog read fails for Node 160."""
    table_name = "genai-powerbi_catalog.default.BIL_PolicyBrand_Id_Xt"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    with pytest.raises(Exception, match="Catalog read failed"):
        read_table(table_name, local_file_path=None)

# ======================================================================
# Tests for: Chunk 226 (Lines 6781-6820)
# ======================================================================

def test_read_table_node_161_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 161."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyBrandType"
    local_path = "/mnt/data/ndw_type/PolicyBrandType.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Type1"), (2, "Type2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "type_name" in result_df.columns


def test_read_table_node_161_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 161."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyBrandType"
    local_path = "/mnt/data/ndw_type/PolicyBrandType.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Type1"), (2, "Type2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "type_name" in result_df.columns


def test_read_table_node_161_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 161."""
    table_name = "genai-powerbi_catalog.ndw_type.PolicyBrandType"
    local_path = "/mnt/data/ndw_type/PolicyBrandType.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 227 (Lines 6811-6850)
# ======================================================================

def test_read_table_node_162_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 162."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5e0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5E0_INCR.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Increment1"), (2, "Increment2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("increment_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "increment_name" in result_df.columns


def test_read_table_node_162_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 162."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5e0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5E0_INCR.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Increment1"), (2, "Increment2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("increment_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "increment_name" in result_df.columns


def test_read_table_node_162_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 162."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5e0_incr"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5E0_INCR.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 228 (Lines 6841-6880)
# ======================================================================

def test_read_table_node_163_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 163."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "DocumentType1"), (2, "DocumentType2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("document_type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "document_type_name" in result_df.columns


def test_read_table_node_163_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 163."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "DocumentType1"), (2, "DocumentType2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("document_type_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "document_type_name" in result_df.columns


def test_read_table_node_163_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 163."""
    table_name = "genai-powerbi_catalog.$NDW_XT.BIL_DocumentType_ID_XT"
    local_path = "/mnt/data/$NDW_XT_BIL_DocumentType_ID_XT.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 229 (Lines 6871-6910)
# ======================================================================

def test_read_table_node_164_catalog_success(spark_session, mocker):
    """Test reading from Unity Catalog successfully for Node 164."""
    table_name = "genai-powerbi_catalog.workdb_billing.tw5td_incr"
    local_path = "/mnt/data/workdb_billing_tw5td_incr.csv"

    # Mock spark.table to return a test DataFrame
    test_data = [(1, "Billing1"), (2, "Billing2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("billing_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.SparkSession.table", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "billing_name" in result_df.columns


def test_read_table_node_164_catalog_failure_local_success(spark_session, mocker):
    """Test fallback to local file when Unity Catalog read fails for Node 164."""
    table_name = "genai-powerbi_catalog.workdb_billing.tw5td_incr"
    local_path = "/mnt/data/workdb_billing_tw5td_incr.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to return a test DataFrame
    test_data = [(1, "Billing1"), (2, "Billing2")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("billing_name", StringType(), True)
    ])
    mock_df = spark_session.createDataFrame(test_data, schema)
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=mock_df)

    result_df = read_table(table_name, local_path)

    assert result_df.count() == 2
    assert "billing_name" in result_df.columns


def test_read_table_node_164_both_fail(spark_session, mocker):
    """Test exception raised when both Unity Catalog and local file read fail for Node 164."""
    table_name = "genai-powerbi_catalog.workdb_billing.tw5td_incr"
    local_path = "/mnt/data/workdb_billing_tw5td_incr.csv"

    # Mock spark.table to raise an exception
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed"))

    # Mock spark.read.format().load to raise an exception
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("Local file read failed"))

    with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog read failed, File error: Local file read failed"):
        read_table(table_name, local_path)

# ======================================================================
# Tests for: Chunk 238 (Lines 7141-7180)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty Step")
        mock_logger.assert_any_call("=== Empty Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")
        mock_logger.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_logger:
            log_df_info(df, "Error Step")
            mock_logger.assert_any_call("=== Error Step ===")
            mock_logger.assert_any_call("   └─ Columns: 2")
            mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 239 (Lines 7171-7210)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result.count() == 2
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result.count() == 2
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/local.csv")
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

def test_read_table_local_file_no_path(spark_session):
    """Test read_table when local file path is None."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        try:
            read_table("catalog.schema.table", None)
        except Exception as e:
            assert str(e) == "Could not read data. Catalog error: Catalog error, File error: None"

def test_read_table_empty_dataframe(spark_session):
    """Test read_table when the resulting DataFrame is empty."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df):
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        assert result.count() == 0
        assert result.columns == ["id", "name"]

# ======================================================================
# Tests for: Chunk 242 (Lines 7261-7300)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Empty Step")
        mock_print.assert_called()

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when DataFrame.show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked error")):
        log_df_info(df, "Error Step")

# ======================================================================
# Tests for: Chunk 243 (Lines 7291-7330)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 246 (Lines 7381-7420)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 249 (Lines 7471-7510)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        df = read_table(table_name, local_file_path)
        assert df.count() == 1
        assert "name" in df.columns

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.test_schema.test_table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 251 (Lines 7531-7570)
# ======================================================================

def test_node_178_read_table_success(spark_session):
    """Test successful reading of table node_178 from Unity Catalog."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5z0_incr"
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = spark_session.table(table_name)
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_node_178_read_table_failure(spark_session):
    """Test failure scenario when reading table node_178 from Unity Catalog."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5z0_incr"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed")):
        try:
            spark_session.table(table_name)
        except Exception as e:
            assert str(e) == "Catalog read failed"

def test_node_178_logging(spark_session):
    """Test logging of DataFrame information for node_178."""
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        df.printSchema()
        df.show(3, truncate=False)
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

# ======================================================================
# Tests for: Chunk 252 (Lines 7561-7600)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test logging of valid DataFrame information."""
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Test Step")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

def test_log_df_info_empty_dataframe(spark_session):
    """Test logging of an empty DataFrame."""
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Test Step")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

def test_log_df_info_show_failure(spark_session):
    """Test logging when DataFrame.show() fails."""
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show failed")):
        log_df_info(df, "Test Step")

# ======================================================================
# Tests for: Chunk 253 (Lines 7591-7630)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test successful reading from Unity Catalog."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = read_table(table_name, "/mnt/data/fallback.csv")
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_read_table_local_file_success(spark_session):
    """Test successful reading from local file."""
    local_file_path = "/mnt/data/fallback.csv"
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrameReader.load", return_value=expected_df):
        result_df = read_table("invalid_table", local_file_path)
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_read_table_both_failures(spark_session):
    """Test failure scenario when both catalog and local file read fail."""
    table_name = "genai_powerbi_catalog.ndw_xt.bil_plancodetype_id_xt"
    local_file_path = "/mnt/data/fallback.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog read failed")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File read failed")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog read failed" in str(e)
            assert "File read failed" in str(e)

# ======================================================================
# Tests for: Chunk 255 (Lines 7651-7690)
# ======================================================================

def test_node_180_read_table_success(spark_session):
    """Test successful reading of table node_180 from Unity Catalog."""
    table_name = "genai-powerbi_catalog.NDW_XT.BIL_PlanCodeType_ID_XT"
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = read_table(table_name, "/mnt/data/fallback.csv")
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_node_180_logging(spark_session):
    """Test logging of DataFrame information for node_180."""
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Source DataFrame Information")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

# ======================================================================
# Tests for: Chunk 256 (Lines 7681-7720)
# ======================================================================

def test_node_181_read_table_success(spark_session):
    """Test successful reading of table node_181 from Unity Catalog."""
    table_name = "genai-powerbi_catalog.ndw_type.PlanCodeType"
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = read_table(table_name, "/mnt/data/fallback.csv")
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_node_181_logging(spark_session):
    """Test logging of DataFrame information for node_181."""
    mock_data = [("1", "PlanA"), ("2", "PlanB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("plan_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Source DataFrame Information")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

# ======================================================================
# Tests for: Chunk 257 (Lines 7711-7750)
# ======================================================================

def test_node_182_read_table_success(spark_session):
    """Test successful reading of table node_182 from Unity Catalog."""
    table_name = "genai_powerbi_catalog.workdb_billing.tw5td_incr"
    mock_data = [("1", "TransactionA"), ("2", "TransactionB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("transaction_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = read_table(table_name, None)
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_node_182_logging(spark_session):
    """Test logging of DataFrame information for node_182."""
    mock_data = [("1", "TransactionA"), ("2", "TransactionB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("transaction_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Source DataFrame Information")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

# ======================================================================
# Tests for: Chunk 258 (Lines 7741-7780)
# ======================================================================

def test_node_183_read_table_success(spark_session):
    """Test successful reading of table node_183 from Unity Catalog."""
    table_name = "genai_powerbi_catalog.default.BIL_TransactionGroupType_Id_XT"
    mock_data = [("1", "GroupA"), ("2", "GroupB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("group_code", StringType(), True)
    ])
    expected_df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=expected_df):
        result_df = read_table(table_name, "/mnt/data/fallback.csv")
        assert result_df.count() == 2
        assert result_df.schema == expected_df.schema

def test_node_183_logging(spark_session):
    """Test logging of DataFrame information for node_183."""
    mock_data = [("1", "GroupA"), ("2", "GroupB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("group_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Source DataFrame Information")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

# ======================================================================
# Tests for: Chunk 259 (Lines 7771-7810)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test logging of valid DataFrame information."""
    mock_data = [("1", "GroupA"), ("2", "GroupB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("group_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Test Step")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

def test_log_df_info_empty_dataframe(spark_session):
    """Test logging of an empty DataFrame."""
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("group_code", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.DataFrame.printSchema") as mock_print_schema, \
         patch("pyspark.sql.DataFrame.show") as mock_show:
        log_df_info(df, "Test Step")
        mock_print_schema.assert_called_once()
        mock_show.assert_called_once_with(3, truncate=False)

def test_log_df_info_show_failure(spark_session):
    """Test logging when DataFrame.show() fails."""
    mock_data = [("1", "GroupA"), ("2", "GroupB")]
    schema = StructType([
        StructField("id", StringType(), True),
        StructField("group_code", StringType(), True)
    ])
    df = spark_session.createDataFrame(mock_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show failed")):
        log_df_info(df, "Test Step")

# ======================================================================
# Tests for: Chunk 260 (Lines 7801-7840)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked Exception")):
        with patch("builtins.print") as mock_print:
            log_df_info(df, "Test Step")
            mock_print.assert_called()

# ======================================================================
# Tests for: Chunk 262 (Lines 7861-7900)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 263 (Lines 7891-7930)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked Exception")):
        with patch("builtins.print") as mock_print:
            log_df_info(df, "Test Step")
            mock_print.assert_called()

# ======================================================================
# Tests for: Chunk 264 (Lines 7921-7960)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 265 (Lines 7951-7990)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 266 (Lines 7981-8020)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info when df.show() raises an exception."""
    test_data = [("Alice", 30), ("Bob", 25)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("age", IntegerType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch.object(df, "show", side_effect=Exception("Mocked Exception")):
        with patch("builtins.print") as mock_print:
            log_df_info(df, "Test Step")
            mock_print.assert_called()

# ======================================================================
# Tests for: Chunk 267 (Lines 8011-8050)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 268 (Lines 8041-8080)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 269 (Lines 8071-8110)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", return_value=mock_df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result == mock_df

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mock_df = MagicMock()
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=mock_df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result == mock_df

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog Error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File Error")):
            with pytest.raises(Exception, match="Could not read data."):
                read_table("catalog.schema.table", "/path/to/local.csv")

# ======================================================================
# Tests for: Chunk 271 (Lines 8131-8170)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info handles exceptions during DataFrame operations."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Mocked error")):
        log_df_info(df, "Test Step")

# ======================================================================
# Tests for: Chunk 272 (Lines 8161-8200)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table successfully reads from Unity Catalog."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table(table_name, local_file_path)
        assert result.count() == 1
        assert "name" in result.columns

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table falls back to local file when catalog read fails."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table(table_name, local_file_path)
        assert result.count() == 1
        assert "name" in result.columns

def test_read_table_both_failures(spark_session):
    """Test read_table raises an exception when both catalog and local file reads fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 273 (Lines 8191-8230)
# ======================================================================

def test_node_190_processing_success(spark_session):
    """Test Node 190 processing with valid data."""
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_190 = read_table(table_name, local_path)
        assert final_df_node_190.count() == 1

def test_node_190_catalog_failure_local_success(spark_session):
    """Test Node 190 falls back to local file when catalog read fails."""
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_190 = read_table(table_name, local_path)
        assert final_df_node_190.count() == 1

def test_node_190_both_failures(spark_session):
    """Test Node 190 raises an exception when both catalog and local file reads fail."""
    table_name = "genai-powerbi_catalog.$WORKDB_BILLING.TW5TD_INCR"
    local_path = "/mnt/data/$WORKDB_BILLING_TW5TD_INCR.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 274 (Lines 8221-8260)
# ======================================================================

def test_node_191_processing_success(spark_session):
    """Test Node 191 processing with valid data."""
    table_name = "genai-powerbi_catalog.default.BIL_SubsystemNtfyCdType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_SubsystemNtfyCdType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_191 = read_table(table_name, local_file_path)
        assert final_df_node_191.count() == 1

def test_node_191_catalog_failure_local_success(spark_session):
    """Test Node 191 falls back to local file when catalog read fails."""
    table_name = "genai-powerbi_catalog.default.BIL_SubsystemNtfyCdType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_SubsystemNtfyCdType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        final_df_node_191 = read_table(table_name, local_file_path)
        assert final_df_node_191.count() == 1

def test_node_191_both_failures(spark_session):
    """Test Node 191 raises an exception when both catalog and local file reads fail."""
    table_name = "genai-powerbi_catalog.default.BIL_SubsystemNtfyCdType_Id_XT"
    local_file_path = "/mnt/data/$NDW_XT_BIL_SubsystemNtfyCdType_Id_XT.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        try:
            read_table(table_name, local_file_path)
        except Exception as e:
            assert "Catalog error" in str(e)
            assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 281 (Lines 8431-8470)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)
    
    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Test Step")
        mock_logger.assert_any_call("=== Test Step ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)
    
    with patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Empty DataFrame Test")
        mock_logger.assert_any_call("=== Empty DataFrame Test ===")
        mock_logger.assert_any_call("   └─ Columns: 2")

def test_log_df_info_show_error(spark_session):
    """Test log_df_info when show() raises an error."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)
    
    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")), \
         patch("logging.Logger.info") as mock_logger:
        log_df_info(df, "Show Error Test")
        mock_logger.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 282 (Lines 8461-8500)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    table_name = "test_catalog.schema.table"
    local_file_path = "/path/to/fallback.csv"
    
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table(table_name, local_file_path)
        assert result.count() == 1
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog fails but local file read succeeds."""
    table_name = "test_catalog.schema.table"
    local_file_path = "/path/to/fallback.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load") as mock_load:
        mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table(table_name, local_file_path)
        assert result.count() == 1
        assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    table_name = "test_catalog.schema.table"
    local_file_path = "/path/to/fallback.csv"
    
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")), \
         patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
        with pytest.raises(Exception, match="Could not read data. Catalog error: Catalog error, File error: File error"):
            read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 294 (Lines 8821-8860)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info handles exceptions during DataFrame operations."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Mocked error")):
        log_df_info(df, "Test Step")

# ======================================================================
# Tests for: Chunk 295 (Lines 8851-8890)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table successfully reads from Unity Catalog."""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        assert result.count() == 1
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table falls back to local file when catalog read fails."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([(2, "Bob")], ["id", "name"])
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            assert result.count() == 1
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table raises an exception when both catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/local.csv")
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 296 (Lines 8881-8920)
# ======================================================================

def test_read_table_node_203_catalog_success(spark_session):
    """Test read_table for Node 203 successfully reads from Unity Catalog."""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table("genai_powerbi_catalog.ndw_xt.bil_nonrenewalcodetype_id_xt", "/path/to/local.csv")
        assert result.count() == 1
        assert result.columns == ["id", "name"]

def test_read_table_node_203_catalog_failure_local_success(spark_session):
    """Test read_table for Node 203 falls back to local file when catalog read fails."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([(2, "Bob")], ["id", "name"])
            result = read_table("genai_powerbi_catalog.ndw_xt.bil_nonrenewalcodetype_id_xt", "/path/to/local.csv")
            assert result.count() == 1
            assert result.columns == ["id", "name"]

def test_read_table_node_203_both_failures(spark_session):
    """Test read_table for Node 203 raises an exception when both catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("genai_powerbi_catalog.ndw_xt.bil_nonrenewalcodetype_id_xt", "/path/to/local.csv")
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 298 (Lines 8941-8980)
# ======================================================================

def test_read_table_node_204_catalog_success(spark_session):
    """Test read_table for Node 204 successfully reads from Unity Catalog."""
    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table("genai_powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/path/to/local.csv")
        assert result.count() == 1
        assert result.columns == ["id", "name"]

def test_read_table_node_204_catalog_failure_local_success(spark_session):
    """Test read_table for Node 204 falls back to local file when catalog read fails."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([(2, "Bob")], ["id", "name"])
            result = read_table("genai_powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/path/to/local.csv")
            assert result.count() == 1
            assert result.columns == ["id", "name"]

def test_read_table_node_204_both_failures(spark_session):
    """Test read_table for Node 204 raises an exception when both catalog and local file reads fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("genai_powerbi_catalog.$WORKDB_BILLING.TZ4A0_INCR", "/path/to/local.csv")
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 305 (Lines 9151-9190)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    try:
        log_df_info(df, "Test Step")
    except Exception as e:
        assert False, f"Unexpected exception raised: {str(e)}"

def test_log_df_info_invalid_dataframe(spark_session):
    """Test log_df_info with an invalid DataFrame (None)."""
    df = None
    try:
        log_df_info(df, "Test Step")
        assert False, "Expected an exception to be raised for None DataFrame"
    except AttributeError:
        pass  # Expected behavior

# ======================================================================
# Tests for: Chunk 306 (Lines 9181-9220)
# ======================================================================

def test_read_table_catalog_success(spark_session, mocker):
    """Test read_table when Unity Catalog read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    result = read_table(table_name, local_file_path)
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session, mocker):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([(1, "Alice")], ["id", "name"]))
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    result = read_table(table_name, local_file_path)
    assert result.count() == 1
    assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session, mocker):
    """Test read_table when both Unity Catalog and local file read fail."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error"))
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    try:
        read_table(table_name, local_file_path)
        assert False, "Expected an exception to be raised"
    except Exception as e:
        assert "Catalog error" in str(e)
        assert "File error" in str(e)

def test_read_table_empty_local_file(spark_session, mocker):
    """Test read_table when local file is empty."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error"))
    mocker.patch("pyspark.sql.DataFrameReader.load", return_value=spark_session.createDataFrame([], StructType([])))
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    result = read_table(table_name, local_file_path)
    assert result.count() == 0

def test_read_table_invalid_table_name(spark_session, mocker):
    """Test read_table with an invalid table name."""
    mocker.patch("pyspark.sql.SparkSession.table", side_effect=Exception("Invalid table name"))
    table_name = None
    local_file_path = "/path/to/local.csv"

    try:
        read_table(table_name, local_file_path)
        assert False, "Expected an exception to be raised"
    except Exception as e:
        assert "Invalid table name" in str(e)

# ======================================================================
# Tests for: Chunk 311 (Lines 9331-9370)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Test Step")
        mock_print.assert_called()

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Empty DataFrame Step")
        mock_print.assert_called()

def test_log_df_info_exception_handling(spark_session):
    """Test log_df_info handles exceptions during DataFrame operations."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Mocked error")):
        log_df_info(df, "Exception Step")

# ======================================================================
# Tests for: Chunk 312 (Lines 9361-9400)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table successfully reads from Unity Catalog."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = read_table(table_name, local_file_path)
        assert result.count() == 1

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table falls back to local file when catalog read fails."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load") as mock_load:
            mock_load.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
            result = read_table(table_name, local_file_path)
            assert result.count() == 1

def test_read_table_both_failures(spark_session):
    """Test read_table raises an exception when both catalog and local file reads fail."""
    table_name = "catalog.schema.table"
    local_file_path = "/path/to/local.csv"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table(table_name, local_file_path)
            except Exception as e:
                assert "Catalog error" in str(e)
                assert "File error" in str(e)

# ======================================================================
# Tests for: Chunk 315 (Lines 9451-9490)
# ======================================================================

def test_read_table_node_216_catalog_success(spark_session):
    """Test Node 216 reads successfully from Unity Catalog."""
    table_name = "genai_powerbi_catalog.workdb_billing.tz4h0_incr"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = spark_session.table(table_name)
        assert result.count() == 1

def test_read_table_node_216_catalog_failure(spark_session):
    """Test Node 216 handles catalog read failure."""
    table_name = "genai_powerbi_catalog.workdb_billing.tz4h0_incr"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        try:
            spark_session.table(table_name)
        except Exception as e:
            assert "Catalog error" in str(e)

# ======================================================================
# Tests for: Chunk 317 (Lines 9511-9550)
# ======================================================================

def test_log_df_info_node_217_valid_dataframe(spark_session):
    """Test log_df_info for Node 217 with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Node 217 Step")
        mock_print.assert_called()

def test_log_df_info_node_217_empty_dataframe(spark_session):
    """Test log_df_info for Node 217 with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("builtins.print") as mock_print:
        log_df_info(df, "Node 217 Empty Step")
        mock_print.assert_called()

# ======================================================================
# Tests for: Chunk 318 (Lines 9541-9580)
# ======================================================================

def test_read_table_node_218_catalog_success(spark_session):
    """Test Node 218 reads successfully from Unity Catalog."""
    table_name = "genai-powerbi_catalog.workdb_billing.tz5h0_incr"

    with patch("pyspark.sql.SparkSession.table") as mock_table:
        mock_table.return_value = spark_session.createDataFrame([(1, "Alice")], ["id", "name"])
        result = spark_session.table(table_name)
        assert result.count() == 1

def test_read_table_node_218_catalog_failure(spark_session):
    """Test Node 218 handles catalog read failure."""
    table_name = "genai-powerbi_catalog.workdb_billing.tz5h0_incr"

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        try:
            spark_session.table(table_name)
        except Exception as e:
            assert "Catalog error" in str(e)

# ======================================================================
# Tests for: Chunk 326 (Lines 9781-9820)
# ======================================================================

def test_log_df_info_valid_dataframe(spark_session):
    """Test log_df_info with a valid DataFrame."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("logging.Logger.info") as mock_info:
        log_df_info(df, "Test Step")
        mock_info.assert_any_call("=== Test Step ===")
        mock_info.assert_any_call("   └─ Columns: 2")
        mock_info.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_empty_dataframe(spark_session):
    """Test log_df_info with an empty DataFrame."""
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame([], schema)

    with patch("logging.Logger.info") as mock_info:
        log_df_info(df, "Empty DataFrame Test")
        mock_info.assert_any_call("=== Empty DataFrame Test ===")
        mock_info.assert_any_call("   └─ Columns: 2")
        mock_info.assert_any_call("   └─ Sample (first 3 rows):")

def test_log_df_info_error_handling(spark_session):
    """Test log_df_info when show() raises an exception."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.DataFrame.show", side_effect=Exception("Show error")):
        with patch("logging.Logger.info") as mock_info:
            log_df_info(df, "Error Handling Test")
            mock_info.assert_any_call("=== Error Handling Test ===")
            mock_info.assert_any_call("   └─ Columns: 2")
            mock_info.assert_any_call("   └─ Could not show data: Show error")

# ======================================================================
# Tests for: Chunk 327 (Lines 9811-9850)
# ======================================================================

def test_read_table_catalog_success(spark_session):
    """Test read_table when Unity Catalog read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", return_value=df) as mock_table:
        result = read_table("catalog.schema.table", "/path/to/local.csv")
        mock_table.assert_called_once_with("catalog.schema.table")
        assert result.count() == 2
        assert result.columns == ["id", "name"]

def test_read_table_catalog_failure_local_success(spark_session):
    """Test read_table when Unity Catalog read fails but local file read succeeds."""
    test_data = [(1, "Alice"), (2, "Bob")]
    schema = StructType([
        StructField("id", IntegerType(), True),
        StructField("name", StringType(), True)
    ])
    df = spark_session.createDataFrame(test_data, schema)

    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", return_value=df) as mock_load:
            result = read_table("catalog.schema.table", "/path/to/local.csv")
            mock_load.assert_called_once_with("/path/to/local.csv")
            assert result.count() == 2
            assert result.columns == ["id", "name"]

def test_read_table_both_failures(spark_session):
    """Test read_table when both Unity Catalog and local file read fail."""
    with patch("pyspark.sql.SparkSession.table", side_effect=Exception("Catalog error")):
        with patch("pyspark.sql.DataFrameReader.load", side_effect=Exception("File error")):
            try:
                read_table("catalog.schema.table", "/path/to/local.csv")
            except Exception as e:
                assert str(e) == "Could not read data. Catalog error: Catalog error, File error: File error"

# ======================================================================
# Tests for: Chunk 330 (Lines 9901-9940)
# ======================================================================

@patch("your_module.read_table")
def test_read_table_valid_source(mock_read_table, spark_session):
    """Test reading data from a valid source table."""
    # Mock the read_table function
    mock_data = [("Alice", 1), ("Bob", 2)]
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame(mock_data, schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai-powerbi_catalog.default.$WORKDB_BILLING.TW5N0_INCR"
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5N0_INCR.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert_df_equality(result_df, mock_df)

@patch("your_module.read_table")
def test_read_table_empty_source(mock_read_table, spark_session):
    """Test reading data when the source table is empty."""
    # Mock the read_table function
    schema = StructType([
        StructField("name", StringType(), True),
        StructField("id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame([], schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai-powerbi_catalog.default.$WORKDB_BILLING.TW5N0_INCR"
    local_file_path = "/mnt/data/$WORKDB_BILLING_TW5N0_INCR.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert result_df.count() == 0

@patch("your_module.read_table")
def test_read_table_invalid_path(mock_read_table):
    """Test reading data with an invalid file path."""
    # Mock the read_table function to raise an exception
    mock_read_table.side_effect = Exception("File not found")

    # Define inputs
    table_name = "genai-powerbi_catalog.default.$WORKDB_BILLING.TW5N0_INCR"
    local_file_path = "/invalid/path.csv"

    # Call the function and validate exception
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 331 (Lines 9931-9970)
# ======================================================================

@patch("your_module.read_table")
def test_read_table_valid_source_node_227(mock_read_table, spark_session):
    """Test reading data from a valid source table for node_227."""
    # Mock the read_table function
    mock_data = [("SystemA", 100), ("SystemB", 200)]
    schema = StructType([
        StructField("system_code", StringType(), True),
        StructField("external_id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame(mock_data, schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai_powerbi_catalog.ndw_xt.bil_externalsystemtype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert_df_equality(result_df, mock_df)

@patch("your_module.read_table")
def test_read_table_empty_source_node_227(mock_read_table, spark_session):
    """Test reading data when the source table is empty for node_227."""
    # Mock the read_table function
    schema = StructType([
        StructField("system_code", StringType(), True),
        StructField("external_id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame([], schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai_powerbi_catalog.ndw_xt.bil_externalsystemtype_id_xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert result_df.count() == 0

@patch("your_module.read_table")
def test_read_table_invalid_path_node_227(mock_read_table):
    """Test reading data with an invalid file path for node_227."""
    # Mock the read_table function to raise an exception
    mock_read_table.side_effect = Exception("File not found")

    # Define inputs
    table_name = "genai_powerbi_catalog.ndw_xt.bil_externalsystemtype_id_xt"
    local_file_path = "/invalid/path.csv"

    # Call the function and validate exception
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)

# ======================================================================
# Tests for: Chunk 332 (Lines 9961-10000)
# ======================================================================

@patch("your_module.read_table")
def test_read_table_valid_source_node_228(mock_read_table, spark_session):
    """Test reading data from a valid source table for node_228."""
    # Mock the read_table function
    mock_data = [("TypeA", 1), ("TypeB", 2)]
    schema = StructType([
        StructField("type_name", StringType(), True),
        StructField("type_id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame(mock_data, schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai-powerbi_catalog.default.BIL_ExternalSystemType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert_df_equality(result_df, mock_df)

@patch("your_module.read_table")
def test_read_table_empty_source_node_228(mock_read_table, spark_session):
    """Test reading data when the source table is empty for node_228."""
    # Mock the read_table function
    schema = StructType([
        StructField("type_name", StringType(), True),
        StructField("type_id", IntegerType(), True)
    ])
    mock_df = spark_session.createDataFrame([], schema)
    mock_read_table.return_value = mock_df

    # Define inputs
    table_name = "genai-powerbi_catalog.default.BIL_ExternalSystemType_Id_Xt"
    local_file_path = "/mnt/data/$NDW_XT_BIL_ExternalSystemType_Id_Xt.csv"

    # Call the function
    result_df = read_table(table_name, local_file_path)

    # Validate the result
    assert result_df.count() == 0

@patch("your_module.read_table")
def test_read_table_invalid_path_node_228(mock_read_table):
    """Test reading data with an invalid file path for node_228."""
    # Mock the read_table function to raise an exception
    mock_read_table.side_effect = Exception("File not found")

    # Define inputs
    table_name = "genai-powerbi_catalog.default.BIL_ExternalSystemType_Id_Xt"
    local_file_path = "/invalid/path.csv"

    # Call the function and validate exception
    with pytest.raises(Exception, match="File not found"):
        read_table(table_name, local_file_path)
```
